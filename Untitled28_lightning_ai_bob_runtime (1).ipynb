## Runtime lane mode — choose before running the runtime

All core cells are compatible with both modes and are tagged **MODE: BOTH**.

- **SINGLE**: one model replica and one model lane. Use this for the one-lane production queue and lower GPU-memory use.
- **MULTI**: multiple independent model replicas and sticky pipeline lanes. The default maximum is three and requires enough GPU memory for every loaded replica.
- **MODE SWITCH**: run the configuration cell once, before Phase 0.5/0.6. To change modes after the model runtime or API has started, restart the kernel and run from the top. Do not rerun only the switch cell against a live server.
- **OPTIONAL** cells are safe in either mode but are not required for production startup.

There are no separate SINGLE-only or MULTI-only implementation cells: preload, generation, cleanup, health, request routing, and API startup all use the selected configuration. This prevents accidentally running the wrong implementation for the selected mode.

````python
# Phase 0.4 — Shared append-only JSON I/O helpers
# MODE: BOTH — safe in single and multi mode
# Every JSON file in OUTPUT_DIR is stored as an append-only record log.
# load_json(filename) returns the latest record's data so old code still works.

OUTPUT_DIR = "./pipeline_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
_json_write_lock = threading.RLock()


def now_iso() -> str:
    from datetime import datetime
    return datetime.utcnow().isoformat() + "Z"


def _json_path(filename: str) -> str:
    return os.path.join(OUTPUT_DIR, filename)


def _empty_json_log(filename: str) -> dict:
    return {"schema_version": "append-log-v1", "filename": filename, "records": []}


def _as_append_log(existing, filename: str) -> dict:
    '''Normalize old/raw JSON files into the append-log shape.'''
    if isinstance(existing, dict) and existing.get("schema_version") == "append-log-v1" and isinstance(existing.get("records"), list):
        return existing
    log = _empty_json_log(filename)
    if existing is not None:
        log["records"].append({
            "index": 0,
            "timestamp": now_iso(),
            "data": existing,
            "migrated_from_raw_json": True,
        })
    return log


def load_json_history(filename: str) -> dict:
    '''Load the full append-only JSON history for filename.'''
    path = _json_path(filename)
    if not os.path.exists(path):
        return _empty_json_log(filename)
    with open(path, "r", encoding="utf-8") as f:
        existing = json.load(f)
    return _as_append_log(existing, filename)


def append_json_record(data, filename: str, meta: Optional[dict] = None) -> dict:
    '''Append a new indexed record to OUTPUT_DIR/filename and return that record.'''
    with _json_write_lock:
        log = load_json_history(filename)
        index = len(log.get("records", []))
        record = {"index": index, "timestamp": now_iso(), "data": data}
        if meta:
            record["meta"] = meta
        log.setdefault("records", []).append(record)
        target = _json_path(filename)
        temporary = f"{target}.{os.getpid()}.{uuid.uuid4().hex}.tmp"
        with open(temporary, "w", encoding="utf-8") as f:
            json.dump(log, f, indent=2, ensure_ascii=False)
            f.flush()
            os.fsync(f.fileno())
        os.replace(temporary, target)
        return record


def save_json(data, filename: str, meta: Optional[dict] = None):
    '''Append data to OUTPUT_DIR/filename instead of overwriting the logical history.'''
    record = append_json_record(data, filename, meta=meta)
    print(f"🧾 Appended {filename} record index={record['index']}")
    return record


def load_json(filename: str):
    '''Return latest record's data from OUTPUT_DIR/filename. None if no records exist.'''
    log = load_json_history(filename)
    records = log.get("records", [])
    if not records:
        return None
    return records[-1].get("data")


def load_json_record(filename: str, index: int):
    '''Return a specific indexed record's data, or None if not found.'''
    log = load_json_history(filename)
    for record in log.get("records", []):
        if record.get("index") == index:
            return record.get("data")
    return None


def extract_json(text: str) -> dict:
    '''Extract the first valid JSON object from a model response, ignoring markdown fences.'''
    match = re.search(r"```(?:json)?\s*(\{.*?\})\s*```", text, re.DOTALL)
    if match:
        json_str = match.group(1)
    else:
        start = text.find("{")
        end = text.rfind("}")
        if start != -1 and end != -1 and end > start:
            json_str = text[start:end + 1]
        else:
            raise ValueError("No JSON object found in response")
    return json.loads(json_str)

print("✅ Append-only JSON helpers ready. OUTPUT_DIR =", os.path.abspath(OUTPUT_DIR))
````

````python
# Phase 0.5 — GPU/RAM cleanup utilities
# MODE: BOTH — safe in single and multi mode
# Policy: keep the shared model loaded; clear only temporary tensors between stages.


_bob_gpu_cleanup_lock = globals().get("_bob_gpu_cleanup_lock", threading.RLock())


def clear_temporary_generation_memory():
    """Clear temporary GPU cache only when no other lane is generating."""
    if int(globals().get("_bob_active_generation_count", 0) or 0) > 0:
        return
    with _bob_gpu_cleanup_lock:
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
            torch.cuda.ipc_collect()
            allocated = torch.cuda.memory_allocated() / 1e9
            reserved = torch.cuda.memory_reserved() / 1e9
            print(f"Temporary memory cleared. Allocated: {allocated:.2f} GB | Reserved: {reserved:.2f} GB")
        else:
            print("CPU run - temporary memory cleared.")


def clear_gpu_memory():
    '''Backward-compatible alias.'''
    clear_temporary_generation_memory()


def clear_stage_context(*names):
    '''Best-effort deletion of named large globals, then clear temp memory.'''
    for name in names:
        if name in globals():
            try:
                del globals()[name]
            except Exception:
                pass
    clear_temporary_generation_memory()


def unload_all_models():
    """Unload every configured model lane in either runtime mode."""
    unload = globals().get("_unload_shared_model")
    if callable(unload):
        unload(all_lanes=True)
    clear_temporary_generation_memory()


print("✅ Cleanup helpers ready.")
````

````python
# Phase 0.6 — Mode-aware LLM runtime [RUN IN SINGLE OR MULTI MODE]
# MODE: BOTH — uses the Phase 0.2A selection
# Each configured lane owns an independent 4-bit model/tokenizer; the request
# pipeline and thread-local workspace context use the same functions in both modes.

SHARED_MODEL_NAME = "Qwen/Qwen2.5-Coder-7B-Instruct"
MODEL_ROOT = Path(os.environ.get("BOB_MODEL_DIR", "./models")).resolve()
MODEL_LOCAL_DIR = MODEL_ROOT / SHARED_MODEL_NAME.rsplit("/", 1)[-1]
if "BOB_RUNTIME_MODE" not in globals() or "BOB_MODEL_LANE_COUNT" not in globals():
    raise RuntimeError("Run the Phase 0.2A MODE SWITCH cell before the model runtime")
_BOB_RUNTIME_INITIALIZED = True

_downloaded_model_path = None
_bob_runtime_state = threading.local()
_bob_model_download_lock = threading.RLock()
_bob_model_load_locks = [threading.RLock() for _ in range(BOB_MODEL_LANE_COUNT)]
_bob_model_lanes = [None for _ in range(BOB_MODEL_LANE_COUNT)]
_bob_tokenizer_lanes = [None for _ in range(BOB_MODEL_LANE_COUNT)]
_bob_generation_state_lock = threading.RLock()
_bob_active_generation_count = 0


def _current_model_lane() -> int:
    lane = int(getattr(_bob_runtime_state, "model_lane", 0) or 0)
    if lane < 0 or lane >= BOB_MODEL_LANE_COUNT:
        raise RuntimeError(f"Invalid model lane {lane}; expected 0..{BOB_MODEL_LANE_COUNT - 1}")
    return lane


def download_shared_model(force_download: bool = False) -> str:
    """Download/resume the shared repository once and return its local path."""
    global _downloaded_model_path
    with _bob_model_download_lock:
        if _downloaded_model_path and not force_download:
            return _downloaded_model_path
        MODEL_LOCAL_DIR.mkdir(parents=True, exist_ok=True)
        print(f"Model repository: {SHARED_MODEL_NAME}")
        print(f"Local model directory: {MODEL_LOCAL_DIR}")
        _downloaded_model_path = snapshot_download(
            repo_id=SHARED_MODEL_NAME,
            local_dir=str(MODEL_LOCAL_DIR),
            token=os.environ.get("HF_TOKEN") or None,
            force_download=force_download,
        )
        print("✅ Model download complete.")
        return _downloaded_model_path


def _load_shared_model(model_lane: Optional[int] = None):
    """Lazily load and return the independent model/tokenizer for one lane."""
    lane = _current_model_lane() if model_lane is None else int(model_lane)
    if lane < 0 or lane >= BOB_MODEL_LANE_COUNT:
        raise RuntimeError(f"Invalid model lane {lane}; expected 0..{BOB_MODEL_LANE_COUNT - 1}")
    with _bob_model_load_locks[lane]:
        if _bob_model_lanes[lane] is not None and _bob_tokenizer_lanes[lane] is not None:
            return _bob_model_lanes[lane], _bob_tokenizer_lanes[lane]
        if not torch.cuda.is_available():
            raise RuntimeError(
                "No CUDA GPU is available. In Lightning AI Studio, stop the Studio, "
                "select a GPU machine, restart it, and rerun the notebook."
            )
        model_path = _downloaded_model_path or download_shared_model()
        print(f"Loading model lane {lane} on {torch.cuda.get_device_name(0)}...")
        bnb_config = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_use_double_quant=True,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_compute_dtype=torch.float16,
        )
        tokenizer = AutoTokenizer.from_pretrained(
            model_path,
            local_files_only=True,
            use_fast=True,
        )
        model = AutoModelForCausalLM.from_pretrained(
            model_path,
            local_files_only=True,
            quantization_config=bnb_config,
            device_map="auto",
            dtype=torch.float16,
            low_cpu_mem_usage=True,
        )
        model.eval()
        if tokenizer.pad_token_id is None:
            tokenizer.pad_token = tokenizer.eos_token
        _bob_model_lanes[lane] = model
        _bob_tokenizer_lanes[lane] = tokenizer
        allocated = torch.cuda.memory_allocated() / (1024 ** 3)
        reserved = torch.cuda.memory_reserved() / (1024 ** 3)
        print(f"✅ Model lane {lane} loaded. GPU memory: {allocated:.2f} GB allocated, {reserved:.2f} GB reserved.")
        return model, tokenizer


def preload_shared_model_lanes() -> None:
    """Load lanes sequentially to avoid concurrent downloads and memory spikes."""
    for lane in range(BOB_MODEL_LANE_COUNT):
        _load_shared_model(lane)


def _unload_shared_model(model_lane: Optional[int] = None, all_lanes: bool = False):
    """Unload the current lane, or every lane when explicitly requested."""
    lanes = range(BOB_MODEL_LANE_COUNT) if all_lanes else [_current_model_lane() if model_lane is None else int(model_lane)]
    for lane in lanes:
        with _bob_model_load_locks[lane]:
            model = _bob_model_lanes[lane]
            tokenizer = _bob_tokenizer_lanes[lane]
            if model is not None:
                print(f"Unloading model lane {lane}...")
                _bob_model_lanes[lane] = None
                _bob_tokenizer_lanes[lane] = None
                del model, tokenizer
    clear_temporary_generation_memory()


def run_shared_generation(
    system_prompt: str,
    user_prompt: str,
    max_new_tokens: int = 2000,
    temperature: float = 0.0,
    do_sample: bool = False,
    top_p: float = 0.95,
) -> str:
    """Generate on the model lane assigned to the current sticky pipeline."""
    lane = _current_model_lane()
    model, tokenizer = _load_shared_model(lane)
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(text, return_tensors="pt")
    input_token_count = int(inputs["input_ids"].shape[-1])
    device = model.get_input_embeddings().weight.device
    inputs = {key: value.to(device) for key, value in inputs.items()}
    generation_kwargs = {
        **inputs,
        "max_new_tokens": max_new_tokens,
        "do_sample": do_sample,
        "pad_token_id": tokenizer.pad_token_id,
        "eos_token_id": tokenizer.eos_token_id,
        "repetition_penalty": 1.0,
        "use_cache": True,
    }
    if do_sample:
        generation_kwargs["temperature"] = temperature
        generation_kwargs["top_p"] = top_p
    global _bob_active_generation_count
    with _bob_generation_state_lock:
        _bob_active_generation_count += 1
    try:
        with torch.inference_mode():
            outputs = model.generate(**generation_kwargs)
    finally:
        with _bob_generation_state_lock:
            _bob_active_generation_count = max(0, _bob_active_generation_count - 1)
    generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    output_token_count = int(generated_tokens.shape[-1])
    raw = tokenizer.decode(generated_tokens, skip_special_tokens=True)
    _bob_runtime_state.usage = {
        "input_tokens": input_token_count,
        "output_tokens": output_token_count,
        "total_tokens": input_token_count + output_token_count,
        "model_lane": lane,
        "runtime_mode": BOB_RUNTIME_MODE,
    }
    del messages, text, inputs, outputs, generated_tokens, generation_kwargs
    clear_temporary_generation_memory()
    return raw


print(f"✅ Lightning {BOB_RUNTIME_MODE}-lane runtime ready ({BOB_MODEL_LANE_COUNT} model lane(s)).")
````

````python
# Phase 1 — Bob IDE payload-backed workspace helpers
# MODE: BOTH — safe in single and multi mode
# Colab now receives workspace data from Bob IDE and must NOT treat a local
# Colab folder as the source of truth. Local WORKSPACE_DIR remains only as a
# fallback for manual notebook testing.

_bob_request_context = globals().get("_bob_request_context", threading.local())


def _bob_context_payload() -> Dict[str, Any]:
    return getattr(_bob_request_context, "payload", {})


def _bob_context_files() -> Dict[str, Optional[str]]:
    return getattr(_bob_request_context, "files", {})


def _bob_context_tree() -> Any:
    return getattr(_bob_request_context, "tree", None)


def _bob_context_active_path() -> str:
    return getattr(_bob_request_context, "active_path", "")


def _bob_context_project() -> str:
    return getattr(_bob_request_context, "project", "sample_project")


if "WORKSPACE_DIR" not in globals() or not globals().get("WORKSPACE_DIR"):
    WORKSPACE_DIR = "./workspace"
    os.makedirs(WORKSPACE_DIR, exist_ok=True)
    print(f"📁 Local fallback workspace set to: {os.path.abspath(WORKSPACE_DIR)}")


def _normalize_path(path: str) -> str:
    path = str(path or "").replace("\\", "/").strip().lstrip("/")
    parts = []
    for part in path.split("/"):
        if part in ("", "."):
            continue
        if part == "..":
            continue
        parts.append(part)
    return "/".join(parts)


def _flatten_workspace_tree(node: Any, prefix: str = "") -> List[str]:
    """Flatten Bob IDE's tree object into visible file paths."""
    files: List[str] = []
    if isinstance(node, list):
        for child in node:
            files.extend(_flatten_workspace_tree(child, prefix))
        return files
    if not isinstance(node, dict):
        return files

    name = str(node.get("name") or "")
    node_type = str(node.get("type") or "")
    path = node.get("path")
    current = _normalize_path(path if path is not None else f"{prefix}/{name}" if prefix and name else name)

    if node_type == "file":
        if current and not current.startswith(".bob/") and "/.bob/" not in current:
            files.append(current)
        return files

    for child in node.get("children") or []:
        files.extend(_flatten_workspace_tree(child, current))
    return files


def workspace_tree_to_text(tree: Any) -> str:
    """Convert Bob IDE workspace_tree JSON into a compact text tree for prompts."""
    if not tree:
        return ""
    paths = sorted(set(_flatten_workspace_tree(tree)))
    if paths:
        return "\n".join(paths)
    try:
        return json.dumps(tree, indent=2, ensure_ascii=False)[:20000]
    except Exception:
        return str(tree)[:20000]


def set_bob_payload_context(payload: Dict[str, Any]) -> Dict[str, Any]:
    """Set request-local workspace state so concurrent pipelines cannot mix context."""
    payload = payload or {}
    files = {
        _normalize_path(k): v
        for k, v in (payload.get("files") or {}).items()
        if _normalize_path(k) and not _normalize_path(k).startswith(".bob/")
    }
    tree = payload.get("workspace_tree") or {}
    active_path = _normalize_path(payload.get("active_path") or payload.get("activePath") or "")
    project = str(payload.get("project") or "sample_project")
    _bob_request_context.payload = payload
    _bob_request_context.files = files
    _bob_request_context.tree = tree
    _bob_request_context.active_path = active_path
    _bob_request_context.project = project
    return {
        "project": project,
        "active_path": active_path,
        "file_count": len(files),
        "tree_count": len(_flatten_workspace_tree(tree)),
    }


def clear_bob_payload_context():
    """Clear only the current request thread's context; model lanes stay loaded."""
    _bob_request_context.payload = {}
    _bob_request_context.files = {}
    _bob_request_context.tree = None
    _bob_request_context.active_path = ""
    _bob_request_context.project = "sample_project"
    clear_temporary_generation_memory()


def list_files() -> List[str]:
    """Return visible relative file paths from the active Bob payload or local fallback."""
    payload_files = sorted(_bob_context_files().keys())
    tree_files = sorted(set(_flatten_workspace_tree(_bob_context_tree()))) if _bob_context_tree() else []
    merged = sorted(set(payload_files + tree_files))
    if merged:
        return [p for p in merged if not p.startswith(".bob/") and "/.bob/" not in p]

    file_list = []
    for root, dirs, files in os.walk(WORKSPACE_DIR):
        dirs[:] = [d for d in dirs if d not in {".git", ".bob", "__pycache__", "node_modules"}]
        for f in files:
            rel = os.path.relpath(os.path.join(root, f), WORKSPACE_DIR).replace("\\", "/")
            if not rel.startswith(".bob/") and "/.bob/" not in rel:
                file_list.append(rel)
    return sorted(file_list)


def read_file(path: str) -> str:
    """Read a file from Bob payload first, then local fallback. Returns '' if missing/deleted."""
    rel = _normalize_path(path)
    if rel in _bob_context_files():
        content = _bob_context_files().get(rel)
        return "" if content is None else str(content)
    full = os.path.join(WORKSPACE_DIR, rel)
    if os.path.isfile(full):
        with open(full, "r", encoding="utf-8") as f:
            return f.read()
    return ""


def save_files(file_dict: Dict[str, Optional[str]]):
    """Local fallback writer only. Do NOT use for Bob IDE/Colab integration."""
    if _bob_context_files():
        raise RuntimeError("Refusing to write files in Colab request mode. Return proposals to Bob IDE instead.")
    for rel_path, content in file_dict.items():
        rel_path = _normalize_path(rel_path)
        full_path = os.path.join(WORKSPACE_DIR, rel_path)
        os.makedirs(os.path.dirname(full_path) or ".", exist_ok=True)
        if content is None:
            if os.path.exists(full_path):
                os.remove(full_path)
                print(f"🗑️ Deleted {rel_path}")
            continue
        if os.path.exists(full_path):
            timestamp = int(time.time())
            backup = f"{full_path}.bak_{timestamp}"
            os.rename(full_path, backup)
            print(f"📁 Backed up {rel_path} → {os.path.basename(backup)}")
        with open(full_path, "w", encoding="utf-8") as f:
            f.write(str(content))
        print(f"✅ Saved {rel_path}")


def get_workspace_tree() -> str:
    """Prompt-friendly workspace file list from Bob payload or local fallback."""
    if _bob_context_tree():
        text = workspace_tree_to_text(_bob_context_tree())
        if text:
            return text
    return "\n".join(list_files())

print("✅ Bob IDE payload workspace helpers ready.")
````

````python
# Phase 1.5 — Proposal/change logging helpers
# MODE: BOTH — safe in single and multi mode
# Colab must return generated files as proposals. It must not write Bob IDE
# project files directly. The app-side JSON worktree owns apply/discard/stage.

import difflib

CHANGELOG_FILE = "changes.json"


def sha256_text(value: Optional[str]) -> str:
    if value is None:
        value = ""
    return hashlib.sha256(str(value).encode("utf-8")).hexdigest()


def read_existing_or_empty(rel_path: str) -> str:
    return read_file(rel_path)


def build_unified_diff(rel_path: str, before: str, after: Optional[str]) -> str:
    after_text = "" if after is None else str(after)
    diff = difflib.unified_diff(
        before.splitlines(keepends=True),
        after_text.splitlines(keepends=True),
        fromfile=f"a/{rel_path}",
        tofile=f"b/{rel_path}",
        lineterm="",
    )
    return "\n".join(diff)


def classify_file_action(before: str, after: Optional[str]) -> str:
    if after is None:
        return "delete"
    if before == "" and str(after) != "":
        return "create"
    if before != str(after):
        return "modify"
    return "unchanged"


def create_proposal_records(file_dict: Dict[str, Optional[str]], run_id: str, review_status: str = "PASS") -> List[dict]:
    """Return indexed proposal metadata for Bob IDE; do not write files."""
    proposals = []
    for local_index, (rel_path, after_content) in enumerate((file_dict or {}).items(), start=1):
        rel_path = _normalize_path(rel_path)
        before_content = read_existing_or_empty(rel_path)
        after_text = None if after_content is None else str(after_content)
        action = classify_file_action(before_content, after_text)
        change_id = f"{run_id}_proposal_{local_index:03d}"
        proposal = {
            "index": local_index,
            "change_id": change_id,
            "run_id": run_id,
            "source": "colab_model",
            "path": rel_path,
            "action": action,
            "status": "proposed" if action != "unchanged" else "unchanged",
            "review_status": review_status,
            "before_hash": sha256_text(before_content),
            "after_hash": sha256_text(after_text),
            "base_hash": sha256_text(before_content),
            "before_size": len(before_content),
            "after_size": 0 if after_text is None else len(after_text),
            "diff": build_unified_diff(rel_path, before_content, after_text),
            "worktree_contract": {
                "app_should_store_as": "proposed",
                "app_should_write_file_now": False,
                "apply_rule": "Bob IDE applies after user approval and hash/review checks",
            },
        }
        proposals.append(proposal)
    return proposals


def create_change_record(run_id: str, user_prompt: str, plan: dict, review: str,
                         file_changes: list, final_status: str) -> dict:
    return {
        "run_id": run_id,
        "timestamp": now_iso(),
        "project": _bob_context_project(),
        "user_prompt": user_prompt,
        "summary": plan.get("summary", ""),
        "final_status": final_status,
        "plan": plan,
        "review": review,
        "files": file_changes,
        "future_worktree_mapping": {
            "branch_name": f"ai/{run_id}",
            "base_ref": "current_workspace_snapshot",
            "patch_format": "unified_diff",
            "apply_strategy": "store as proposed .bob changes; apply only after user approval",
            "commit_message": plan.get("summary", "AI generated changes"),
            "mcp_tools_later": [
                "worktree.status", "worktree.get_diff", "worktree.apply_change",
                "worktree.discard_change", "worktree.stage_change", "worktree.create_snapshot",
                "model.plan", "model.run_agent",
            ],
        },
    }


def save_files_with_change_log(file_dict: Dict[str, Optional[str]], run_id: str, review_status: str = "PASS") -> List[dict]:
    """Compatibility wrapper: now records proposals only and never writes files."""
    proposals = create_proposal_records(file_dict, run_id=run_id, review_status=review_status)
    save_json({"run_id": run_id, "proposals": proposals}, "proposals.json", meta={"stage": "proposal_builder", "run_id": run_id})
    print(f"🧾 Built {len(proposals)} proposal record(s). No files written by Colab.")
    return proposals

print("✅ Proposal/change logging helpers ready. Colab will not write project files.")
````

````python
# Phase 3 — Planner
# MODE: BOTH — safe in single and multi mode
PLANNER_MODEL = SHARED_MODEL_NAME

PLANNER_SYSTEM = '''You are a senior software architect. You NEVER write code.

Your task is to understand the user request and create a detailed plan.

You will be given:
- The user request.
- The Bob IDE workspace tree.
- Optional file contents supplied by Bob IDE.

CRITICAL RULES:
- If a workspace tree is present, you have already performed a workspace scan. Do NOT request_workspace.
- If file contents are present, use only those contents as source-of-truth.
- If you need specific files, set output_mode = "request_files" and list them in files_needed.
- If you have enough context, set output_mode = "ready_for_coder".
- The coder_prompt must be concrete and implementation-oriented.
- Return only a valid JSON object using the exact normalized Plan schema.

Schema:
{
  "task_type": "bugfix | feature | refactor | docs | investigation",
  "summary": "short summary",
  "confidence": 0.0,
  "output_mode": "request_workspace | request_files | request_docs | ready_for_coder",
  "need_workspace_scan": false,
  "need_file_contents": false,
  "need_documentation": false,
  "investigation_points": [],
  "possible_causes": [],
  "required_context": [],
  "files_needed": [],
  "documentation_needed": [],
  "tool_calls": [],
  "reasoning_steps": [],
  "coder_prompt": ""
}'''


def _load_planner_model():
    return _load_shared_model()


def _unload_planner_model():
    print("Planner stage complete. Shared model kept loaded.")
    clear_temporary_generation_memory()


def _run_planner_model(user_prompt: str, workspace_tree: str = "", file_contents: str = "") -> dict:
    """Run planner using a fresh prompt on the shared model."""
    full_prompt = f"User Request:\n{user_prompt}\n"
    if _bob_context_project():
        full_prompt += f"\nProject: {_bob_context_project()}\n"
    if _bob_context_active_path():
        full_prompt += f"Active file: {_bob_context_active_path()}\n"
    if workspace_tree:
        full_prompt += f"\n## Workspace tree from Bob IDE:\n{workspace_tree}\n"
        full_prompt += "\nIMPORTANT: workspace_tree is already supplied. Do not request_workspace.\n"
    if file_contents:
        full_prompt += f"\n## File contents supplied by Bob IDE:\n{file_contents}\n"

    raw = run_shared_generation(PLANNER_SYSTEM, full_prompt, max_new_tokens=2000, temperature=0.0, do_sample=False)
    try:
        data = normalize_plan(extract_json(raw))
    except Exception as e:
        print(f"⚠️ JSON parsing error: {e}")
        data = normalize_plan({
            "task_type": "investigation",
            "summary": "Planner output could not be parsed",
            "confidence": 0.1,
            "output_mode": "ready_for_coder" if workspace_tree else "request_workspace",
            "coder_prompt": f"Make a conservative best-effort implementation for: {user_prompt}",
        })
    del full_prompt, raw
    clear_temporary_generation_memory()
    return data


def planner_step(user_prompt: str, workspace_tree: str = "", file_contents: str = "",
                 force_rerun: bool = False, force_ready: bool = False,
                 run_id: Optional[str] = None) -> dict:
    """Step 1: produce a normalized Plan. Appends to plan.json."""
    if not force_rerun:
        cached = load_json("plan.json")
        if cached:
            print("📂 Using latest cached plan.json record.")
            cached_plan = cached.get("plan", cached)
            if force_ready and cached_plan.get("output_mode") != "ready_for_coder":
                cached_plan["output_mode"] = "ready_for_coder"
                if not cached_plan.get("coder_prompt"):
                    cached_plan["coder_prompt"] = f"Create the necessary files for: {user_prompt}"
            return normalize_plan(cached_plan)

    plan = _run_planner_model(user_prompt, workspace_tree, file_contents)
    if force_ready and plan.get("output_mode") != "ready_for_coder":
        plan["output_mode"] = "ready_for_coder"
        if not plan.get("coder_prompt"):
            plan["coder_prompt"] = f"Create the necessary files for: {user_prompt}"
    save_json({"run_id": run_id, "project": _bob_context_project(), "plan": plan}, "plan.json", meta={"stage": "planner", "run_id": run_id})
    print("✅ Plan appended to plan.json")
    return plan

print("✅ Planner ready for Bob IDE payloads.")
````

````python
# Phase 4 — Context builder
# MODE: BOTH (OPTIONAL) — safe in single and multi mode
def build_context_step(plan: dict, force_rerun: bool = False, run_id: Optional[str] = None) -> str:
    """Build coder context from Bob IDE payload files. Does not scan hidden .bob."""
    if plan is None:
        print("⚠️ Warning: plan is None. Returning empty context.")
        return ""

    if not force_rerun:
        cached = load_json("context.json")
        if cached:
            print("📂 Using cached context.")
            return cached.get("context", "")

    all_visible_files = list_files()
    files_to_read = [_normalize_path(p) for p in (plan.get("files_needed") or [])]

    # If planner did not ask for specific files, use the backend-capped payload set.
    # Bob IDE places active file first in payload when available.
    if not files_to_read:
        files_to_read = list(_bob_context_files().keys()) or all_visible_files
        print(f"DEBUG: No files_needed, using supplied payload/local files: {files_to_read}")

    # Ensure active file is first if present.
    if _bob_context_active_path() and _bob_context_active_path() in all_visible_files and _bob_context_active_path() not in files_to_read:
        files_to_read.insert(0, _bob_context_active_path())

    context = "Project: " + _bob_context_project() + "\n"
    context += "Active file: " + (_bob_context_active_path() or "") + "\n\n"
    context += "Workspace files:\n" + "\n".join(all_visible_files) + "\n\n"

    total_chars = 0
    max_chars = 520_000  # keep close to Bob IDE's ~500 KB context cap
    for file in files_to_read:
        file = _normalize_path(file)
        if not file or file.startswith(".bob/") or "/.bob/" in file:
            continue
        content = read_file(file)
        block = f"===== FILE: {file} =====\n{content}\n\n" if content else f"===== FILE: {file} NOT SUPPLIED OR EMPTY =====\n\n"
        if total_chars + len(block) > max_chars:
            context += f"===== CONTEXT TRUNCATED BEFORE: {file} =====\n"
            break
        context += block
        total_chars += len(block)

    print(f"DEBUG: build_context finished, total length: {len(context)}")
    save_json({"run_id": run_id, "project": _bob_context_project(), "context": context}, "context.json", meta={"stage": "context_builder", "run_id": run_id})
    print("✅ Context appended to context.json")
    return context

print("✅ Context builder ready for Bob IDE payloads.")
````

````python
# Phase 6 — Parser for coder output
# MODE: BOTH (OPTIONAL) — safe in single and multi mode
def parse_coder_output(output: str, expected_files: Optional[List[str]] = None) -> Dict[str, str]:
    '''
    Extract file paths and contents from the coder's response.

    Looks for headers like ### "path/to/file.py" or FILE: path/to/file.py
    immediately followed by a markdown code block. Falls back to inferring a
    name from the code block's language, from a path-like string just before
    the block, from expected_files, or finally a generic name.

    Returns dict {file_path: content}.
    '''
    block_pattern = r'''
        (?:
            (?:\#+\s*`([^`]+)`)|
            (?:FILE:\s*`?([^`\n]+)`?)
        )?\s*
        ```(\w+)?\s*\n
        (.*?)
        ```
    '''
    matches = list(re.finditer(block_pattern, output, re.DOTALL | re.VERBOSE))
    if not matches:
        return {}

    blocks = []
    for match in matches:
        header_path1 = match.group(1)
        header_path2 = match.group(2)
        lang = match.group(3)
        content = match.group(4).strip()
        blocks.append({
            "header": header_path1 or header_path2,
            "lang": lang,
            "content": content,
            "start": match.start(),
            "end": match.end(),
        })

    files: Dict[str, str] = {}
    for block in blocks:
        if block["header"]:
            files[block["header"]] = block["content"]

    unnamed = [b for b in blocks if not b["header"]]
    if unnamed:
        if expected_files and len(unnamed) == len(expected_files):
            for idx, block in enumerate(unnamed):
                files[expected_files[idx]] = block["content"]
            return files

        ext_map = {
            "python": ".py", "html": ".html", "css": ".css",
            "javascript": ".js", "json": ".json", "xml": ".xml",
            "yaml": ".yml", "markdown": ".md", "bash": ".sh",
            "shell": ".sh", "sql": ".sql",
        }
        for idx, block in enumerate(unnamed):
            file_name = None
            if block["lang"]:
                ext = ext_map.get(block["lang"].lower(), f'.{block["lang"].lower()}')
                file_name = "app" + ext if len(blocks) == 1 else f"file_{idx + 1}{ext}"
            else:
                start = max(0, block["start"] - 300)
                preceding = output[start:block["start"]]
                path_match = re.search(r"([^\s\n]+\.(?:py|html|css|js|json|xml|yml|md|sh|sql))", preceding)
                if path_match:
                    file_name = path_match.group(1).strip()
                elif expected_files:
                    used = set(files.keys())
                    for exp in expected_files:
                        if exp not in used:
                            file_name = exp
                            break
                if not file_name:
                    file_name = f"file_{idx + 1}.txt"
            files[file_name] = block["content"]

    if not files:
        single = re.search(r"```(?:\w+)?\s*\n(.*?)```", output, re.DOTALL)
        if single:
            files["app.py"] = single.group(1).strip()

    print(f"Parser: extracted {len(files)} file(s): {list(files.keys())}")
    return files

print("✅ Parser ready.")
````

````python
# Phase 7 — Improved Reviewer: deterministic checks + semantic LLM review
# MODE: BOTH — safe in single and multi mode
#
# Improvements:
# - Treats plan.files_needed as required files, not an exclusive whitelist.
# - Allows necessary supporting files such as templates, CSS, tests, and configs.
# - Performs deterministic parsing and Python syntax validation.
# - Detects missing referenced templates and local modules.
# - Gives the model structured evidence instead of only raw generated text.
# - Requires actionable, file-specific feedback.
# - Normalizes malformed reviewer output into the existing PASS/FAIL contract.
# - Remains compatible with reviewer_step(), /review, and run_agent().

import ast
import json
import re
from pathlib import PurePosixPath
from typing import Any, Dict, List, Optional, Set, Tuple


REVIEWER_SYSTEM = r'''
You are the senior code-review agent for Bob IDE.

You are reviewing proposed full-file replacements generated by another model.
The files have NOT been written to the real workspace. They will become
proposed worktree changes only after review and user approval.

Your job is to determine whether the proposal is reasonably correct,
complete, safe, internally consistent, and aligned with the user's request.

IMPORTANT REVIEW RULES

1. PLAN FILES ARE MINIMUM REQUIREMENTS, NOT A WHITELIST
   `plan.files_needed` identifies files the planner expected to inspect,
   create, or modify. It is not automatically an exclusive list.

   Do not fail merely because the coder added a supporting file.

   Supporting files are valid when they are reasonably required by the
   implementation, including:
   - HTML templates
   - CSS or JavaScript assets
   - tests
   - configuration files
   - package manifests
   - migration files
   - documentation
   - helper modules
   - deployment files explicitly justified by the request

2. REVIEW THE COMPLETE IMPLEMENTATION
   Evaluate all generated files together. Understand imports, routes,
   templates, configuration, function calls, data flow, and dependencies
   across files.

3. DISTINGUISH BLOCKING AND NON-BLOCKING ISSUES
   Return FAIL only for issues that would likely:
   - prevent the code from parsing, importing, starting, or functioning;
   - violate a core user requirement;
   - omit an essential implementation file;
   - introduce a material security vulnerability;
   - destroy or overwrite unrelated functionality;
   - contain fabricated, contradictory, or unusable code;
   - perform direct or unsafe filesystem mutation contrary to the worktree contract.

   Minor style concerns, optional improvements, comments, formatting,
   dependency pinning preferences, or speculative edge cases should not
   cause FAIL.

4. DO NOT INVENT ERRORS
   Only report an issue when supported by the supplied code or deterministic
   evidence. Do not claim an import is missing when it is present. Do not
   claim syntax failure without identifying the exact file and construct.

5. FULL-FILE REPLACEMENT AWARENESS
   Each generated file is intended to be its complete proposed content.
   Check whether an existing-file replacement appears to remove necessary
   behavior, but do not assume unseen code exists.

6. SECURITY REVIEW
   Identify material issues such as:
   - hard-coded real credentials or tokens;
   - command injection;
   - unsafe eval/exec;
   - path traversal;
   - disabled authentication or authorization;
   - insecure deserialization;
   - destructive filesystem operations;
   - obvious SQL injection;
   - exposing secrets in responses.

   Placeholder development secrets should normally be WARNINGS unless the
   request clearly requires production-ready security.

7. BE PRACTICAL
   Judge whether the implementation satisfies the request, not whether it is
   the most advanced possible architecture.

INTERNAL REVIEW PROCESS

Before deciding, reason through:
A. What does the user want?
B. What does the plan require?
C. Which files were generated?
D. Why does each generated file exist?
E. Do references between files resolve?
F. Are there syntax, runtime, logic, security, or completeness problems?
G. Which problems are genuinely blocking?

OUTPUT CONTRACT

Return exactly one JSON object and no Markdown fences:

{
  "verdict": "PASS" or "FAIL",
  "summary": "One concise overall assessment.",
  "blocking_issues": [
    {
      "file": "relative/path.ext or <project>",
      "category": "syntax|runtime|logic|security|completeness|contract",
      "issue": "Specific verified problem.",
      "evidence": "Relevant symbol, statement, or relationship.",
      "fix": "Concrete correction the coder should make."
    }
  ],
  "warnings": [
    {
      "file": "relative/path.ext or <project>",
      "issue": "Non-blocking concern.",
      "suggestion": "Optional improvement."
    }
  ],
  "reviewed_files": ["path1", "path2"],
  "requirement_checks": [
    {
      "requirement": "Requirement being checked.",
      "status": "satisfied|unsatisfied|uncertain",
      "evidence": "Why."
    }
  ]
}

Verdict rules:
- PASS when blocking_issues is empty.
- FAIL when one or more blocking_issues are present.
- Never use a warning as the sole reason for FAIL.
'''


# ---------------------------------------------------------------------------
# General helpers
# ---------------------------------------------------------------------------

def _load_reviewer_model():
    return _load_shared_model()


def _unload_reviewer_model():
    print("Reviewer stage complete. Shared model kept loaded.")
    clear_temporary_generation_memory()


def _review_normalize_path(path: Any) -> str:
    """Normalize a generated relative path without touching the filesystem."""
    value = str(path or "").strip().replace("\\", "/")

    while value.startswith("./"):
        value = value[2:]

    value = re.sub(r"/+", "/", value)
    return value.strip("/")


def _review_unique_strings(values: Any) -> List[str]:
    """Return an ordered, normalized, duplicate-free string list."""
    if not isinstance(values, (list, tuple, set)):
        return []

    result: List[str] = []
    seen: Set[str] = set()

    for value in values:
        normalized = _review_normalize_path(value)
        if normalized and normalized not in seen:
            seen.add(normalized)
            result.append(normalized)

    return result


def _review_file_language(path: str) -> str:
    suffix = PurePosixPath(path).suffix.lower()

    return {
        ".py": "python",
        ".pyi": "python",
        ".js": "javascript",
        ".jsx": "javascript",
        ".mjs": "javascript",
        ".cjs": "javascript",
        ".ts": "typescript",
        ".tsx": "typescript",
        ".html": "html",
        ".htm": "html",
        ".css": "css",
        ".scss": "scss",
        ".sass": "sass",
        ".json": "json",
        ".yaml": "yaml",
        ".yml": "yaml",
        ".toml": "toml",
        ".md": "markdown",
        ".sql": "sql",
        ".sh": "shell",
        ".ps1": "powershell",
        ".java": "java",
        ".go": "go",
        ".rs": "rust",
        ".cpp": "cpp",
        ".cc": "cpp",
        ".c": "c",
        ".h": "c",
        ".hpp": "cpp",
    }.get(suffix, "text")


def _review_json_object_from_text(raw: str) -> Optional[dict]:
    """
    Extract the first valid JSON object from model output.

    Handles:
    - clean JSON
    - ```json fenced JSON
    - explanatory text before or after JSON
    """
    text = str(raw or "").strip()
    if not text:
        return None

    text = re.sub(r"^\s*```(?:json)?\s*", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\s*```\s*$", "", text)

    try:
        value = json.loads(text)
        return value if isinstance(value, dict) else None
    except Exception:
        pass

    decoder = json.JSONDecoder()

    for index, character in enumerate(text):
        if character != "{":
            continue

        try:
            value, _ = decoder.raw_decode(text[index:])
            if isinstance(value, dict):
                return value
        except Exception:
            continue

    return None


# ---------------------------------------------------------------------------
# Parsing generated files
# ---------------------------------------------------------------------------

def _review_parse_files(
    code: str,
    expected_files: Optional[List[str]] = None,
) -> Tuple[Dict[str, Optional[str]], List[str]]:
    """
    Parse the coder output and return:
        (generated_files, parser_problems)

    The primary parser remains parse_coder_output() so this reviewer stays
    compatible with the rest of the notebook.
    """
    expected_files = _review_unique_strings(expected_files or [])
    problems: List[str] = []

    try:
        parsed = parse_coder_output(code or "", expected_files=expected_files)
    except Exception as exc:
        return {}, [f"parse_coder_output failed: {type(exc).__name__}: {exc}"]

    if not isinstance(parsed, dict):
        return {}, [
            "parse_coder_output returned a non-dictionary result: "
            f"{type(parsed).__name__}"
        ]

    normalized: Dict[str, Optional[str]] = {}

    for raw_path, raw_content in parsed.items():
        path = _review_normalize_path(raw_path)

        if not path:
            problems.append("The coder output contains an empty file path.")
            continue

        if path in normalized:
            problems.append(f"The coder defined the file more than once: {path}")
            continue

        if raw_content is None:
            normalized[path] = None
        else:
            normalized[path] = str(raw_content)

    return normalized, problems


# ---------------------------------------------------------------------------
# Deterministic validation
# ---------------------------------------------------------------------------

def _review_python_imports(tree: ast.AST) -> List[str]:
    imports: List[str] = []

    for node in ast.walk(tree):
        if isinstance(node, ast.Import):
            imports.extend(alias.name for alias in node.names)

        elif isinstance(node, ast.ImportFrom):
            if node.module:
                prefix = "." * int(getattr(node, "level", 0) or 0)
                imports.append(prefix + node.module)
            elif getattr(node, "level", 0):
                imports.append("." * int(node.level))

    return imports


def _review_python_template_references(tree: ast.AST) -> List[str]:
    references: List[str] = []

    for node in ast.walk(tree):
        if not isinstance(node, ast.Call):
            continue

        function_name = None

        if isinstance(node.func, ast.Name):
            function_name = node.func.id
        elif isinstance(node.func, ast.Attribute):
            function_name = node.func.attr

        if function_name not in {
            "render_template",
            "get_template",
            "TemplateResponse",
        }:
            continue

        if not node.args:
            continue

        first_arg = node.args[0]

        if isinstance(first_arg, ast.Constant) and isinstance(first_arg.value, str):
            references.append(first_arg.value)

    return references


def _review_python_dangerous_calls(tree: ast.AST) -> List[dict]:
    """Find a small set of high-confidence dangerous operations."""
    issues: List[dict] = []

    for node in ast.walk(tree):
        if not isinstance(node, ast.Call):
            continue

        call_name = ""

        if isinstance(node.func, ast.Name):
            call_name = node.func.id
        elif isinstance(node.func, ast.Attribute):
            parts = []
            current = node.func

            while isinstance(current, ast.Attribute):
                parts.append(current.attr)
                current = current.value

            if isinstance(current, ast.Name):
                parts.append(current.id)

            call_name = ".".join(reversed(parts))

        line_number = int(getattr(node, "lineno", 0) or 0)

        if call_name in {"eval", "exec"}:
            issues.append({
                "category": "security",
                "issue": f"Use of {call_name}() can execute untrusted code.",
                "evidence": f"{call_name} call at line {line_number}",
            })

        if call_name in {"os.system", "subprocess.call", "subprocess.run", "subprocess.Popen"}:
            shell_true = any(
                keyword.arg == "shell"
                and isinstance(keyword.value, ast.Constant)
                and keyword.value.value is True
                for keyword in node.keywords
            )

            if call_name == "os.system" or shell_true:
                issues.append({
                    "category": "security",
                    "issue": "Shell command execution requires careful input validation.",
                    "evidence": f"{call_name} at line {line_number}",
                })

        if call_name in {
            "shutil.rmtree",
            "os.remove",
            "os.unlink",
            "Path.unlink",
        }:
            issues.append({
                "category": "contract",
                "issue": "Generated application code contains a destructive filesystem operation.",
                "evidence": f"{call_name} at line {line_number}",
            })

    return issues


def _review_expected_template_candidates(template_name: str) -> Set[str]:
    clean = _review_normalize_path(template_name)

    return {
        clean,
        _review_normalize_path(f"templates/{clean}"),
        _review_normalize_path(f"app/templates/{clean}"),
        _review_normalize_path(f"src/templates/{clean}"),
    }


def _review_local_module_candidates(import_name: str) -> Set[str]:
    """
    Convert an import into likely local file paths.

    Example:
        package.helpers -> package/helpers.py
        .helpers        -> helpers.py
    """
    clean = str(import_name or "").lstrip(".").strip()
    if not clean:
        return set()

    module_path = clean.replace(".", "/")

    return {
        f"{module_path}.py",
        f"{module_path}/__init__.py",
    }


def _review_is_probably_local_import(import_name: str, generated_paths: Set[str]) -> bool:
    """
    Only consider an import probably local if the generated proposal includes
    a matching module or package. This avoids incorrectly flagging third-party
    libraries as missing generated files.
    """
    candidates = _review_local_module_candidates(import_name)
    return any(candidate in generated_paths for candidate in candidates)


def _review_validate_files(
    plan: dict,
    generated_files: Dict[str, Optional[str]],
    parser_problems: List[str],
) -> dict:
    """
    Produce deterministic evidence for the LLM reviewer.

    These checks focus on high-confidence facts. The LLM performs the broader
    semantic and requirement-level review.
    """
    expected_files = _review_unique_strings(plan.get("files_needed") or [])
    generated_paths = list(generated_files.keys())
    generated_path_set = set(generated_paths)

    blocking_issues: List[dict] = []
    warnings: List[dict] = []
    file_analysis: Dict[str, dict] = {}
    referenced_templates: List[dict] = []
    local_import_relationships: List[dict] = []

    # Parser-level failures
    for problem in parser_problems:
        blocking_issues.append({
            "file": "<coder-output>",
            "category": "contract",
            "issue": problem,
            "evidence": "The generated response could not be reliably converted into file proposals.",
            "fix": (
                "Return each complete file using the required "
                "### `relative/path` heading and fenced content format."
            ),
        })

    if not generated_files:
        blocking_issues.append({
            "file": "<coder-output>",
            "category": "completeness",
            "issue": "No generated files were extracted from the coder output.",
            "evidence": "The parsed generated-file dictionary is empty.",
            "fix": "Return at least one complete file proposal in the required format.",
        })

    # Required files are a minimum requirement, not an exclusive list.
    missing_expected = [
        path for path in expected_files
        if path not in generated_path_set
    ]

    for path in missing_expected:
        blocking_issues.append({
            "file": path,
            "category": "completeness",
            "issue": "A file explicitly required by the plan is missing.",
            "evidence": (
                f"{path!r} appears in plan.files_needed but was not extracted "
                "from the coder output."
            ),
            "fix": f"Return the complete proposed content for {path}.",
        })

    # Analyze every generated file.
    for path, content in generated_files.items():
        language = _review_file_language(path)

        analysis = {
            "language": language,
            "size_chars": 0 if content is None else len(content),
            "is_deletion": content is None,
            "syntax_valid": None,
            "syntax_error": None,
            "imports": [],
            "template_references": [],
        }

        file_analysis[path] = analysis

        if content is None:
            warnings.append({
                "file": path,
                "issue": "The proposal deletes this file.",
                "suggestion": (
                    "Confirm that deletion is required and does not remove "
                    "unrelated project behavior."
                ),
            })
            continue

        if not content.strip():
            blocking_issues.append({
                "file": path,
                "category": "completeness",
                "issue": "The generated file content is empty.",
                "evidence": "The proposed full-file replacement contains no code or text.",
                "fix": "Return the complete intended content or mark the file for deletion.",
            })
            continue

        suffix = PurePosixPath(path).suffix.lower()

        # Python syntax and relationship analysis
        if suffix in {".py", ".pyi"}:
            try:
                tree = ast.parse(content, filename=path)
                analysis["syntax_valid"] = True
            except SyntaxError as exc:
                analysis["syntax_valid"] = False
                analysis["syntax_error"] = {
                    "line": exc.lineno,
                    "offset": exc.offset,
                    "message": exc.msg,
                    "text": (exc.text or "").strip(),
                }

                blocking_issues.append({
                    "file": path,
                    "category": "syntax",
                    "issue": f"Python syntax error: {exc.msg}",
                    "evidence": (
                        f"Line {exc.lineno}, column {exc.offset}: "
                        f"{(exc.text or '').strip()}"
                    ),
                    "fix": "Correct the syntax and return the complete valid file.",
                })

                tree = None

            if tree is not None:
                imports = _review_python_imports(tree)
                template_refs = _review_python_template_references(tree)

                analysis["imports"] = imports
                analysis["template_references"] = template_refs

                for dangerous in _review_python_dangerous_calls(tree):
                    warnings.append({
                        "file": path,
                        "issue": dangerous["issue"],
                        "suggestion": (
                            "Verify inputs are trusted and replace the operation "
                            "with a safer API where possible. Treat as blocking "
                            "if user-controlled data reaches it."
                        ),
                    })

                for template_name in template_refs:
                    candidates = _review_expected_template_candidates(template_name)
                    present = sorted(candidates.intersection(generated_path_set))

                    referenced_templates.append({
                        "source_file": path,
                        "template": template_name,
                        "candidate_paths": sorted(candidates),
                        "generated_matches": present,
                    })

                    # Do not always fail here because the template may already
                    # exist in the supplied workspace and may not need changing.
                    if not present:
                        warnings.append({
                            "file": path,
                            "issue": (
                                f"The code references template {template_name!r}, "
                                "but no matching template is included in the proposal."
                            ),
                            "suggestion": (
                                "Confirm that the template already exists in the "
                                "workspace. If it does not, generate it."
                            ),
                        })

                for import_name in imports:
                    candidates = _review_local_module_candidates(import_name)
                    present = sorted(candidates.intersection(generated_path_set))

                    if present or _review_is_probably_local_import(
                        import_name,
                        generated_path_set,
                    ):
                        local_import_relationships.append({
                            "source_file": path,
                            "import": import_name,
                            "candidate_paths": sorted(candidates),
                            "generated_matches": present,
                        })

        # JSON syntax
        elif suffix == ".json":
            try:
                json.loads(content)
                analysis["syntax_valid"] = True
            except json.JSONDecodeError as exc:
                analysis["syntax_valid"] = False
                analysis["syntax_error"] = {
                    "line": exc.lineno,
                    "column": exc.colno,
                    "message": exc.msg,
                }

                blocking_issues.append({
                    "file": path,
                    "category": "syntax",
                    "issue": f"Invalid JSON: {exc.msg}",
                    "evidence": f"Line {exc.lineno}, column {exc.colno}",
                    "fix": "Return syntactically valid JSON.",
                })

        # Basic extension/content mismatch detection
        if suffix in {".html", ".htm"}:
            suspicious_python = (
                re.search(r"^\s*(from|import)\s+\w+", content, re.MULTILINE)
                and "<html" not in content.lower()
                and "<!doctype" not in content.lower()
            )

            if suspicious_python:
                blocking_issues.append({
                    "file": path,
                    "category": "contract",
                    "issue": "The HTML file appears to contain Python source code.",
                    "evidence": "Python import syntax was found without recognizable HTML markup.",
                    "fix": "Place the content in the correct file or return valid HTML.",
                })

        elif suffix in {".py", ".pyi"}:
            suspicious_html = (
                ("<!doctype html" in content.lower() or "<html" in content.lower())
                and not re.search(
                    r"(['\"]{3}).*?<html",
                    content,
                    flags=re.IGNORECASE | re.DOTALL,
                )
            )

            if suspicious_html:
                blocking_issues.append({
                    "file": path,
                    "category": "contract",
                    "issue": "The Python file appears to contain raw HTML as its main content.",
                    "evidence": "An HTML document marker was found in a .py file.",
                    "fix": "Move the HTML into a template/static file or wrap it correctly as a Python string.",
                })

    extra_files = [
        path for path in generated_paths
        if path not in expected_files
    ]

    return {
        "expected_files": expected_files,
        "generated_files": generated_paths,
        "missing_expected_files": missing_expected,
        "additional_supporting_files": extra_files,
        "important_note": (
            "Additional supporting files are allowed. They must not be treated "
            "as errors merely because they are absent from plan.files_needed."
        ),
        "deterministic_blocking_issues": blocking_issues,
        "deterministic_warnings": warnings,
        "file_analysis": file_analysis,
        "referenced_templates": referenced_templates,
        "local_import_relationships": local_import_relationships,
    }


# ---------------------------------------------------------------------------
# Prompt construction
# ---------------------------------------------------------------------------

def _review_compact_file_bundle(
    generated_files: Dict[str, Optional[str]],
    per_file_limit: int = 18000,
    total_limit: int = 70000,
) -> str:
    """
    Build a readable file bundle while avoiding an unexpectedly huge prompt.

    For normal Bob proposals, complete files are included. Very large files are
    represented by a head/tail excerpt and clearly marked as truncated.
    """
    sections: List[str] = []
    used = 0

    for path, content in generated_files.items():
        language = _review_file_language(path)

        if content is None:
            body = "<DELETE FILE>"
            truncated = False
        else:
            text = str(content)
            truncated = len(text) > per_file_limit

            if truncated:
                head_size = per_file_limit // 2
                tail_size = per_file_limit - head_size
                body = (
                    text[:head_size]
                    + "\n\n... <CONTENT TRUNCATED FOR REVIEW PROMPT> ...\n\n"
                    + text[-tail_size:]
                )
            else:
                body = text

        section = (
            f"\n===== FILE: {path} =====\n"
            f"LANGUAGE: {language}\n"
            f"TRUNCATED: {str(truncated).lower()}\n"
            f"```{language}\n{body}\n```\n"
        )

        if used + len(section) > total_limit:
            sections.append(
                "\n===== REMAINING FILES OMITTED FROM MODEL PROMPT =====\n"
                "The deterministic report still contains their metadata.\n"
            )
            break

        sections.append(section)
        used += len(section)

    return "".join(sections)


def _review_requirement_list(plan: dict) -> List[str]:
    requirements: List[str] = []

    summary = str(plan.get("summary") or "").strip()
    coder_prompt = str(plan.get("coder_prompt") or "").strip()

    if summary:
        requirements.append(summary)

    acceptance = plan.get("acceptance_criteria") or []
    if isinstance(acceptance, list):
        requirements.extend(
            str(item).strip()
            for item in acceptance
            if str(item).strip()
        )

    if coder_prompt and coder_prompt not in requirements:
        requirements.append(coder_prompt)

    # Avoid sending duplicated historical reviewer feedback as a requirement.
    cleaned: List[str] = []
    seen: Set[str] = set()

    for requirement in requirements:
        requirement = re.split(
            r"\n\s*Reviewer feedback\. Fix these issues:",
            requirement,
            maxsplit=1,
            flags=re.IGNORECASE,
        )[0].strip()

        if requirement and requirement not in seen:
            seen.add(requirement)
            cleaned.append(requirement)

    return cleaned


# ---------------------------------------------------------------------------
# Reviewer result normalization
# ---------------------------------------------------------------------------

def _review_issue_to_text(issue: Any) -> str:
    if not isinstance(issue, dict):
        return f"- {str(issue).strip()}"

    file_path = str(issue.get("file") or "<project>")
    category = str(issue.get("category") or "issue").upper()
    description = str(issue.get("issue") or "Unspecified problem").strip()
    evidence = str(issue.get("evidence") or "").strip()
    fix = str(issue.get("fix") or "").strip()

    lines = [f"- [{category}] {file_path}: {description}"]

    if evidence:
        lines.append(f"  Evidence: {evidence}")

    if fix:
        lines.append(f"  Fix: {fix}")

    return "\n".join(lines)


def _review_warning_to_text(warning: Any) -> str:
    if not isinstance(warning, dict):
        return f"- {str(warning).strip()}"

    file_path = str(warning.get("file") or "<project>")
    issue = str(warning.get("issue") or "Unspecified warning").strip()
    suggestion = str(warning.get("suggestion") or "").strip()

    text = f"- {file_path}: {issue}"

    if suggestion:
        text += f"\n  Suggestion: {suggestion}"

    return text


def _review_normalize_result(
    raw: str,
    deterministic_report: dict,
) -> str:
    """
    Convert model JSON into the notebook's existing text contract:

        PASS
        ...

    or:

        FAIL
        ...

    Deterministic blocking errors cannot be overridden by the model.
    """
    parsed = _review_json_object_from_text(raw)

    deterministic_blocking = list(
        deterministic_report.get("deterministic_blocking_issues") or []
    )
    deterministic_warnings = list(
        deterministic_report.get("deterministic_warnings") or []
    )

    if parsed is None:
        # Compatibility fallback if the model ignores the JSON instruction but
        # still returns the old PASS/FAIL format.
        stripped = str(raw or "").strip()
        upper = stripped.upper()

        if deterministic_blocking:
            issues_text = "\n".join(
                _review_issue_to_text(issue)
                for issue in deterministic_blocking
            )
            return (
                "FAIL\n"
                "Deterministic validation found blocking issues:\n"
                f"{issues_text}"
            )

        if upper.startswith("PASS"):
            return stripped

        if upper.startswith("FAIL"):
            return stripped

        return (
            "FAIL\n"
            "The reviewer returned an invalid response format and its verdict "
            "could not be verified.\n"
            f"Raw reviewer response:\n{stripped[:3000]}"
        )

    model_blocking = parsed.get("blocking_issues")
    if not isinstance(model_blocking, list):
        model_blocking = []

    model_warnings = parsed.get("warnings")
    if not isinstance(model_warnings, list):
        model_warnings = []

    all_blocking = deterministic_blocking + model_blocking
    all_warnings = deterministic_warnings + model_warnings

    verdict = "FAIL" if all_blocking else "PASS"
    summary = str(parsed.get("summary") or "").strip()

    reviewed_files = parsed.get("reviewed_files")
    if not isinstance(reviewed_files, list):
        reviewed_files = deterministic_report.get("generated_files") or []

    requirement_checks = parsed.get("requirement_checks")
    if not isinstance(requirement_checks, list):
        requirement_checks = []

    output_lines = [verdict]

    if summary:
        output_lines.append(summary)
    elif verdict == "PASS":
        output_lines.append(
            "The proposal satisfies the reviewed requirements and no blocking "
            "issues were verified."
        )
    else:
        output_lines.append(
            "The proposal contains one or more verified blocking issues."
        )

    if all_blocking:
        output_lines.append("\nBlocking issues:")
        output_lines.extend(
            _review_issue_to_text(issue)
            for issue in all_blocking
        )

    if requirement_checks:
        output_lines.append("\nRequirement checks:")

        for check in requirement_checks:
            if not isinstance(check, dict):
                continue

            requirement = str(
                check.get("requirement") or "Unnamed requirement"
            ).strip()
            status = str(check.get("status") or "uncertain").upper()
            evidence = str(check.get("evidence") or "").strip()

            line = f"- [{status}] {requirement}"
            if evidence:
                line += f" — {evidence}"

            output_lines.append(line)

    if all_warnings:
        output_lines.append("\nWarnings:")
        output_lines.extend(
            _review_warning_to_text(warning)
            for warning in all_warnings
        )

    if reviewed_files:
        normalized_reviewed = [
            _review_normalize_path(path)
            for path in reviewed_files
            if _review_normalize_path(path)
        ]

        if normalized_reviewed:
            output_lines.append(
                "\nReviewed files: " + ", ".join(normalized_reviewed)
            )

    return "\n".join(output_lines).strip()


# ---------------------------------------------------------------------------
# Shared-model reviewer
# ---------------------------------------------------------------------------

def _run_reviewer_model(plan: dict, code: str) -> str:
    """Run deterministic validation followed by semantic model review."""
    normalized_plan = dict(plan or {})
    expected_files = _review_unique_strings(
        normalized_plan.get("files_needed") or []
    )

    generated_files, parser_problems = _review_parse_files(
        code,
        expected_files=expected_files,
    )

    deterministic_report = _review_validate_files(
        normalized_plan,
        generated_files,
        parser_problems,
    )

    requirements = _review_requirement_list(normalized_plan)
    file_bundle = _review_compact_file_bundle(generated_files)

    prompt = f'''
PROJECT:
{_bob_context_project()}

USER/PLAN REQUIREMENTS:
{json.dumps(requirements, indent=2, ensure_ascii=False)}

PLAN JSON:
{json.dumps(normalized_plan, indent=2, ensure_ascii=False)}

DETERMINISTIC REVIEW REPORT:
{json.dumps(deterministic_report, indent=2, ensure_ascii=False)}

GENERATED FILE CONTENTS:
{file_bundle}

Review instructions:

1. Review the proposal as one implementation, not as unrelated snippets.
2. Use the deterministic report as evidence, but independently inspect the code.
3. `files_needed` is a minimum expected set, not an exclusive whitelist.
4. Additional supporting files are allowed when justified by the implementation.
5. Do not fail only because a template, test, config, CSS, JavaScript, helper,
   package manifest, or deployment file was added.
6. Determine whether referenced files may reasonably already exist in the
   workspace before treating their absence from this proposal as blocking.
7. Only put verified release-blocking defects in blocking_issues.
8. Put minor concerns and optional improvements in warnings.
9. Check actual imports, definitions, routes, functions, templates, variable
   names, and cross-file references before making a claim.
10. Return exactly the required JSON object.
'''

    raw = run_shared_generation(
        REVIEWER_SYSTEM,
        prompt,
        max_new_tokens=2600,
        temperature=0.0,
        do_sample=False,
    )

    normalized_review = _review_normalize_result(
        raw,
        deterministic_report,
    )

    del prompt
    del file_bundle
    del raw
    clear_temporary_generation_memory()

    return normalized_review


def reviewer_step(
    plan: dict,
    code: str,
    force_rerun: bool = False,
    run_id: Optional[str] = None,
) -> str:
    """
    Step 4: review generated code and append the result to review.json.

    The return format remains compatible with the current orchestrator:
        PASS\\n...
        FAIL\\n...
    """
    if not force_rerun:
        cached = load_json("review.json")

        if cached:
            print("📂 Using latest cached review.json record.")
            return str(cached.get("review") or "")

    review = _run_reviewer_model(plan, code)

    save_json(
        {
            "run_id": run_id,
            "project": _bob_context_project(),
            "review": review,
            "reviewer_version": "semantic-deterministic-v2",
        },
        "review.json",
        meta={
            "stage": "reviewer",
            "run_id": run_id,
            "reviewer_version": "semantic-deterministic-v2",
        },
    )

    print("✅ Improved review appended to review.json")
    return review


print("✅ Improved reviewer agent ready for Bob IDE.")
````

````python
# Phase 7.1 — Reviewer accuracy patch (drop-in replacement)
# MODE: BOTH — safe in single and multi mode
# Run this cell AFTER the existing Phase 7 reviewer cell.
# It replaces only the semantic review / verdict logic and keeps the existing
# parser + deterministic validation helpers.

import ast
import json
import re
from pathlib import PurePosixPath
from typing import Any, Dict, List, Optional, Set


REVIEWER_SYSTEM = r'''
You are the senior code-review agent for Bob IDE.

You are reviewing COMPLETE proposed file contents generated by a coder model.
The proposal has not yet been applied to the user's real workspace.

Your job is not to judge whether the code "looks reasonable". Your job is to
verify, using concrete evidence from the supplied plan, deterministic report,
existing-file context, and generated files, that the coder actually implemented
the requested change correctly.

========================
MANDATORY REVIEW METHOD
========================

Before deciding PASS or FAIL, perform all of the following checks internally:

1. REQUIREMENT COVERAGE
   - Check EVERY numbered requirement supplied in REVIEW REQUIREMENTS.
   - For each requirement, identify the exact generated file(s), symbol(s),
     route(s), component(s), function(s), configuration, or statements that
     satisfy it.
   - Never mark a requirement satisfied merely because the coder claims it is.
   - If an explicit requirement is not implemented, mark it unsatisfied.
   - If the supplied evidence is genuinely insufficient to verify an explicit
     requirement, mark it uncertain. Do not silently assume success.

2. FILE-BY-FILE REVIEW
   - Review EVERY generated file, not just the first or most important file.
   - Confirm the file is relevant to the implementation.
   - Check syntax evidence, imports, names, calls, state/data flow, error paths,
     API contracts, and obvious runtime problems.
   - Review all generated files together as one implementation.

3. CROSS-FILE CONSISTENCY
   Verify relationships such as:
   - imported symbols actually being defined/exported;
   - route names and client URLs matching;
   - function signatures matching their callers;
   - template names matching referenced templates;
   - DOM ids/classes/selectors matching JavaScript/CSS usage;
   - configuration keys/environment variables matching their consumers;
   - request/response shapes matching between frontend/backend;
   - renamed symbols being updated at their call sites;
   - new dependencies being represented where required.

4. FULL-FILE REPLACEMENT SAFETY
   Generated files are complete replacements. Use EXISTING FILE CHANGE CONTEXT
   to look for suspicious removal of existing behavior, public symbols, routes,
   or substantial file content. Do not fail merely because code was removed,
   but fail if the proposal clearly destroys unrelated required behavior.

5. RUNTIME / LOGIC CHECK
   Mentally trace the important execution path created or modified by the
   proposal. Look for undefined variables, wrong branches, unreachable logic,
   incorrect async usage, wrong data shapes, bad return values, missing awaits,
   mismatched arguments, missing initialization, and likely crashes.

6. SECURITY / CONTRACT CHECK
   Check for material issues such as real hard-coded credentials, path
   traversal, command injection, unsafe eval/exec, disabled auth, insecure
   deserialization, destructive filesystem operations, or violations of Bob's
   proposal/worktree contract.

========================
VERDICT POLICY
========================

FAIL when ANY of these is true:
- deterministic validation reports a blocking issue;
- an explicit review requirement is unsatisfied;
- an explicit review requirement cannot be verified because an essential part
  of the implementation is missing;
- a generated file was not actually reviewed;
- the implementation contains a likely syntax/import/runtime/logic failure;
- a material cross-file contract is inconsistent;
- a material security or worktree-contract issue exists;
- a full-file replacement clearly removes unrelated required behavior.

PASS only when:
- every explicit requirement is satisfied with concrete evidence;
- every generated file has been reviewed;
- there are no verified blocking issues;
- cross-file relationships are internally consistent enough to reasonably run.

Minor style concerns, naming preferences, optional refactors, speculative edge
cases, and non-essential polish belong in warnings and MUST NOT cause FAIL.

Do not invent defects. Every blocking issue must cite concrete evidence from the
supplied proposal or deterministic report.

========================
OUTPUT CONTRACT
========================

Return EXACTLY one JSON object and no Markdown fences:

{
  "verdict": "PASS" or "FAIL",
  "summary": "Concise overall assessment.",
  "blocking_issues": [
    {
      "file": "relative/path.ext or <project>",
      "category": "syntax|runtime|logic|security|completeness|contract|regression",
      "issue": "Specific verified blocking problem.",
      "evidence": "Concrete code/symbol/relationship evidence.",
      "fix": "Concrete correction the coder should make."
    }
  ],
  "warnings": [
    {
      "file": "relative/path.ext or <project>",
      "issue": "Non-blocking concern.",
      "suggestion": "Optional improvement."
    }
  ],
  "reviewed_files": ["every/generated/file.ext"],
  "requirement_checks": [
    {
      "id": "R1",
      "requirement": "Exact requirement being checked.",
      "status": "satisfied|unsatisfied|uncertain",
      "evidence": "Specific file + symbol/statement proving the status."
    }
  ],
  "cross_file_checks": [
    {
      "relationship": "What was checked.",
      "status": "satisfied|broken|uncertain",
      "evidence": "Concrete evidence."
    }
  ]
}

Important:
- requirement_checks MUST contain exactly one entry for every supplied R# item.
- reviewed_files MUST contain every generated file path you inspected.
- A requirement marked unsatisfied MUST correspond to a blocking issue.
- A cross-file check marked broken MUST correspond to a blocking issue.
'''


def _review_requirements_structured(plan: dict) -> List[dict]:
    """Create explicit numbered requirements that the reviewer must prove."""
    raw_requirements: List[str] = []

    summary = str(plan.get("summary") or "").strip()
    if summary:
        raw_requirements.append(summary)

    acceptance = plan.get("acceptance_criteria") or []
    if isinstance(acceptance, list):
        raw_requirements.extend(
            str(item).strip()
            for item in acceptance
            if str(item).strip()
        )

    coder_prompt = str(plan.get("coder_prompt") or "").strip()

    if coder_prompt:
        coder_prompt = re.split(
            r"\n\s*Reviewer feedback\. Fix these issues:",
            coder_prompt,
            maxsplit=1,
            flags=re.IGNORECASE,
        )[0].strip()

        if coder_prompt:
            raw_requirements.append(coder_prompt)

    steps = plan.get("steps") or []

    if isinstance(steps, list):
        for step in steps:
            if isinstance(step, dict):
                text = str(
                    step.get("description")
                    or step.get("task")
                    or step.get("action")
                    or ""
                ).strip()
            else:
                text = str(step).strip()

            if text:
                raw_requirements.append(text)

    cleaned: List[str] = []
    seen: Set[str] = set()

    for item in raw_requirements:
        normalized = re.sub(r"\s+", " ", item).strip()
        key = normalized.casefold()

        if normalized and key not in seen:
            seen.add(key)
            cleaned.append(normalized)

    return [
        {
            "id": f"R{index}",
            "requirement": text,
        }
        for index, text in enumerate(cleaned, start=1)
    ]


def _review_python_public_symbols(content: str) -> dict:
    """
    Extract high-signal top-level Python symbols.

    This helps detect accidental removals when the coder returns a complete
    replacement for an existing file.
    """
    result = {
        "functions": [],
        "classes": [],
        "imports": [],
        "routes": [],
    }

    try:
        tree = ast.parse(content)
    except Exception:
        return result

    for node in tree.body:

        if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)):
            result["functions"].append(node.name)

            for decorator in node.decorator_list:

                if (
                    isinstance(decorator, ast.Call)
                    and isinstance(decorator.func, ast.Attribute)
                ):
                    if decorator.func.attr in {
                        "route",
                        "get",
                        "post",
                        "put",
                        "patch",
                        "delete",
                    }:
                        if (
                            decorator.args
                            and isinstance(decorator.args[0], ast.Constant)
                        ):
                            value = decorator.args[0].value

                            if isinstance(value, str):
                                result["routes"].append(value)

        elif isinstance(node, ast.ClassDef):
            result["classes"].append(node.name)

        elif isinstance(node, ast.Import):
            result["imports"].extend(
                alias.name for alias in node.names
            )

        elif isinstance(node, ast.ImportFrom) and node.module:
            result["imports"].append(node.module)

    for key in result:
        result[key] = sorted(set(result[key]))

    return result


def _review_existing_change_context(
    generated_files: Dict[str, Optional[str]],
    max_excerpt_chars: int = 5000,
) -> dict:
    """
    Compare each generated file against the current workspace version.

    This is especially useful because the coder output represents COMPLETE
    proposed file contents rather than patches.
    """
    context: Dict[str, dict] = {}

    for path, proposed in generated_files.items():

        try:
            existing = read_file(path)
        except Exception:
            existing = ""

        existing = "" if existing is None else str(existing)
        proposed_text = "" if proposed is None else str(proposed)

        entry = {
            "exists_in_workspace": bool(existing),
            "existing_chars": len(existing),
            "proposed_chars": (
                0 if proposed is None else len(proposed_text)
            ),
            "action": (
                "delete"
                if proposed is None
                else "create"
                if not existing
                else "unchanged"
                if existing == proposed_text
                else "modify"
            ),
            "size_change_ratio": None,
            "removed_python_symbols": {},
            "added_python_symbols": {},
        }

        if existing:
            entry["size_change_ratio"] = round(
                (
                    len(proposed_text)
                    - len(existing)
                )
                / max(1, len(existing)),
                3,
            )

        if (
            path.endswith((".py", ".pyi"))
            and existing
            and proposed is not None
        ):
            before = _review_python_public_symbols(existing)
            after = _review_python_public_symbols(proposed_text)

            for kind in (
                "functions",
                "classes",
                "routes",
                "imports",
            ):
                removed = sorted(
                    set(before[kind]) - set(after[kind])
                )

                added = sorted(
                    set(after[kind]) - set(before[kind])
                )

                if removed:
                    entry["removed_python_symbols"][kind] = removed

                if added:
                    entry["added_python_symbols"][kind] = added

        if (
            existing
            and proposed is not None
            and existing != proposed_text
        ):
            try:
                diff = build_unified_diff(
                    path,
                    existing,
                    proposed_text,
                )
            except Exception:
                diff = ""

            if diff:

                if len(diff) > max_excerpt_chars:
                    half = max_excerpt_chars // 2

                    diff = (
                        diff[:half]
                        + "\n... <DIFF TRUNCATED> ...\n"
                        + diff[-half:]
                    )

                entry["diff_excerpt"] = diff

        context[path] = entry

    return context


def _review_model_issue(
    file_path: str,
    category: str,
    issue: str,
    evidence: str,
    fix: str,
) -> dict:

    return {
        "file": file_path,
        "category": category,
        "issue": issue,
        "evidence": evidence,
        "fix": fix,
    }


def _review_normalize_result(
    raw: str,
    deterministic_report: dict,
    requirements: Optional[List[dict]] = None,
) -> str:
    """
    Strictly normalize the reviewer result.

    Important difference from v2:

    An unsatisfied requirement, missing requirement check, broken relationship,
    or missing reviewed-file coverage CANNOT accidentally result in PASS.
    """
    requirements = requirements or []

    parsed = _review_json_object_from_text(raw)

    deterministic_blocking = list(
        deterministic_report.get(
            "deterministic_blocking_issues"
        )
        or []
    )

    deterministic_warnings = list(
        deterministic_report.get(
            "deterministic_warnings"
        )
        or []
    )

    if parsed is None:

        if deterministic_blocking:
            return (
                "FAIL\n"
                "Deterministic validation found blocking issues:\n"
                + "\n".join(
                    _review_issue_to_text(issue)
                    for issue in deterministic_blocking
                )
            )

        return (
            "FAIL\n"
            "Reviewer returned an invalid response format, so the "
            "review cannot be trusted.\n"
            "Raw reviewer response:\n"
            + str(raw or "")[:3000]
        )

    model_blocking = parsed.get("blocking_issues")

    if not isinstance(model_blocking, list):
        model_blocking = []

    model_warnings = parsed.get("warnings")

    if not isinstance(model_warnings, list):
        model_warnings = []

    # ==========================================================
    # FILE REVIEW COVERAGE
    # ==========================================================

    generated_files = [
        _review_normalize_path(path)
        for path in (
            deterministic_report.get("generated_files")
            or []
        )
        if _review_normalize_path(path)
    ]

    reviewed_files_raw = parsed.get("reviewed_files")

    reviewed_files = []

    if isinstance(reviewed_files_raw, list):
        reviewed_files = [
            _review_normalize_path(path)
            for path in reviewed_files_raw
            if _review_normalize_path(path)
        ]

    missing_review_coverage = sorted(
        set(generated_files)
        - set(reviewed_files)
    )

    if missing_review_coverage:
        model_blocking.append(
            _review_model_issue(
                "<reviewer>",
                "contract",
                (
                    "The reviewer did not confirm inspection "
                    "of every generated file."
                ),
                (
                    "Missing reviewed_files coverage: "
                    + ", ".join(missing_review_coverage)
                ),
                (
                    "Review every generated file and include "
                    "each path in reviewed_files."
                ),
            )
        )

    # ==========================================================
    # REQUIREMENT COVERAGE
    # ==========================================================

    requirement_checks = parsed.get(
        "requirement_checks"
    )

    if not isinstance(requirement_checks, list):
        requirement_checks = []

    expected_by_id = {
        str(item.get("id")): str(
            item.get("requirement") or ""
        )
        for item in requirements
        if isinstance(item, dict)
        and item.get("id")
    }

    checks_by_id: Dict[str, dict] = {}

    for check in requirement_checks:

        if not isinstance(check, dict):
            continue

        rid = str(
            check.get("id") or ""
        ).strip()

        if rid and rid not in checks_by_id:
            checks_by_id[rid] = check

    missing_requirement_checks = [
        rid
        for rid in expected_by_id
        if rid not in checks_by_id
    ]

    if missing_requirement_checks:
        model_blocking.append(
            _review_model_issue(
                "<reviewer>",
                "contract",
                (
                    "The reviewer failed to evaluate "
                    "every explicit requirement."
                ),
                (
                    "Missing requirement checks: "
                    + ", ".join(
                        missing_requirement_checks
                    )
                ),
                (
                    "Return exactly one requirement_checks "
                    "entry for every supplied R# requirement."
                ),
            )
        )

    for rid, requirement_text in expected_by_id.items():

        check = checks_by_id.get(rid)

        if not check:
            continue

        status = str(
            check.get("status")
            or "uncertain"
        ).strip().lower()

        evidence = str(
            check.get("evidence")
            or ""
        ).strip()

        if status == "unsatisfied":

            model_blocking.append(
                _review_model_issue(
                    "<project>",
                    "completeness",
                    (
                        f"Explicit requirement {rid} "
                        f"is not satisfied: "
                        f"{requirement_text}"
                    ),
                    (
                        evidence
                        or "Reviewer marked the "
                        "requirement unsatisfied."
                    ),
                    (
                        "Update the implementation so "
                        "this requirement is concretely "
                        "satisfied."
                    ),
                )
            )

        elif status == "uncertain":

            model_blocking.append(
                _review_model_issue(
                    "<project>",
                    "completeness",
                    (
                        f"Explicit requirement {rid} "
                        f"could not be verified: "
                        f"{requirement_text}"
                    ),
                    (
                        evidence
                        or "Reviewer supplied insufficient "
                        "evidence to verify the requirement."
                    ),
                    (
                        "Provide or correct the implementation "
                        "so the requirement can be verified "
                        "from the proposal."
                    ),
                )
            )

        elif status != "satisfied":

            model_blocking.append(
                _review_model_issue(
                    "<reviewer>",
                    "contract",
                    (
                        f"Requirement {rid} has an "
                        f"invalid review status: "
                        f"{status!r}."
                    ),
                    (
                        evidence
                        or (
                            "Expected satisfied, "
                            "unsatisfied, or uncertain."
                        )
                    ),
                    (
                        "Return a valid status and "
                        "concrete evidence."
                    ),
                )
            )

        elif not evidence:

            model_blocking.append(
                _review_model_issue(
                    "<reviewer>",
                    "contract",
                    (
                        f"Requirement {rid} was marked "
                        "satisfied without evidence."
                    ),
                    requirement_text,
                    (
                        "Cite the generated file and "
                        "symbol/statement that satisfies "
                        "the requirement."
                    ),
                )
            )

    # ==========================================================
    # CROSS-FILE CHECKS
    # ==========================================================

    cross_file_checks = parsed.get(
        "cross_file_checks"
    )

    if not isinstance(cross_file_checks, list):
        cross_file_checks = []

    for check in cross_file_checks:

        if not isinstance(check, dict):
            continue

        status = str(
            check.get("status")
            or "uncertain"
        ).strip().lower()

        relationship = str(
            check.get("relationship")
            or "Cross-file relationship"
        ).strip()

        evidence = str(
            check.get("evidence")
            or ""
        ).strip()

        if status == "broken":

            model_blocking.append(
                _review_model_issue(
                    "<project>",
                    "contract",
                    (
                        "Broken cross-file relationship: "
                        + relationship
                    ),
                    (
                        evidence
                        or (
                            "Reviewer marked this "
                            "relationship broken."
                        )
                    ),
                    (
                        "Make the referenced files, "
                        "symbols, and contracts consistent."
                    ),
                )
            )

    # ==========================================================
    # DE-DUPLICATE BLOCKING ISSUES
    # ==========================================================

    all_blocking_raw = (
        deterministic_blocking
        + model_blocking
    )

    all_blocking: List[dict] = []
    seen_issue_keys: Set[str] = set()

    for issue in all_blocking_raw:

        if isinstance(issue, dict):
            key = "|".join(
                [
                    str(
                        issue.get("file")
                        or ""
                    ),
                    str(
                        issue.get("category")
                        or ""
                    ),
                    str(
                        issue.get("issue")
                        or ""
                    ),
                ]
            ).casefold()

        else:
            key = str(issue).casefold()

        if key not in seen_issue_keys:
            seen_issue_keys.add(key)
            all_blocking.append(issue)

    all_warnings = (
        deterministic_warnings
        + model_warnings
    )

    verdict = (
        "FAIL"
        if all_blocking
        else "PASS"
    )

    summary = str(
        parsed.get("summary")
        or ""
    ).strip()

    output_lines = [verdict]

    output_lines.append(
        summary
        or (
            (
                "All explicit requirements were verified "
                "and no blocking issues were found."
            )
            if verdict == "PASS"
            else (
                "The proposal contains one or more "
                "verified blocking issues."
            )
        )
    )

    if all_blocking:

        output_lines.append(
            "\nBlocking issues:"
        )

        output_lines.extend(
            _review_issue_to_text(issue)
            for issue in all_blocking
        )

    if requirement_checks:

        output_lines.append(
            "\nRequirement checks:"
        )

        for check in requirement_checks:

            if not isinstance(check, dict):
                continue

            rid = str(
                check.get("id")
                or "?"
            ).strip()

            requirement = str(
                check.get("requirement")
                or expected_by_id.get(rid)
                or "Unnamed requirement"
            ).strip()

            status = str(
                check.get("status")
                or "uncertain"
            ).upper()

            evidence = str(
                check.get("evidence")
                or ""
            ).strip()

            line = (
                f"- [{rid} / {status}] "
                f"{requirement}"
            )

            if evidence:
                line += (
                    f" — {evidence}"
                )

            output_lines.append(line)

    if cross_file_checks:

        output_lines.append(
            "\nCross-file checks:"
        )

        for check in cross_file_checks:

            if not isinstance(check, dict):
                continue

            relationship = str(
                check.get("relationship")
                or "Cross-file relationship"
            ).strip()

            status = str(
                check.get("status")
                or "uncertain"
            ).upper()

            evidence = str(
                check.get("evidence")
                or ""
            ).strip()

            line = (
                f"- [{status}] "
                f"{relationship}"
            )

            if evidence:
                line += (
                    f" — {evidence}"
                )

            output_lines.append(line)

    if all_warnings:

        output_lines.append(
            "\nWarnings:"
        )

        output_lines.extend(
            _review_warning_to_text(warning)
            for warning in all_warnings
        )

    if reviewed_files:

        output_lines.append(
            "\nReviewed files: "
            + ", ".join(reviewed_files)
        )

    return "\n".join(
        output_lines
    ).strip()


def _run_reviewer_model(
    plan: dict,
    code: str,
) -> str:
    """
    Evidence-driven semantic reviewer.

    Combines:
    - deterministic checks
    - explicit requirements
    - current workspace comparison
    - complete generated file contents
    - strict output verification
    """
    normalized_plan = dict(
        plan or {}
    )

    expected_files = _review_unique_strings(
        normalized_plan.get(
            "files_needed"
        )
        or []
    )

    generated_files, parser_problems = (
        _review_parse_files(
            code,
            expected_files=expected_files,
        )
    )

    deterministic_report = (
        _review_validate_files(
            normalized_plan,
            generated_files,
            parser_problems,
        )
    )

    requirements = (
        _review_requirements_structured(
            normalized_plan
        )
    )

    existing_change_context = (
        _review_existing_change_context(
            generated_files
        )
    )

    file_bundle = (
        _review_compact_file_bundle(
            generated_files,
            per_file_limit=22000,
            total_limit=90000,
        )
    )

    prompt = f'''
PROJECT:
{_bob_context_project()}

REVIEW REQUIREMENTS — CHECK EVERY R# ITEM:
{json.dumps(
    requirements,
    indent=2,
    ensure_ascii=False
)}

PLAN JSON:
{json.dumps(
    normalized_plan,
    indent=2,
    ensure_ascii=False
)}

DETERMINISTIC REVIEW REPORT:
{json.dumps(
    deterministic_report,
    indent=2,
    ensure_ascii=False
)}

EXISTING FILE CHANGE CONTEXT:
{json.dumps(
    existing_change_context,
    indent=2,
    ensure_ascii=False
)}

GENERATED COMPLETE FILE PROPOSALS:
{file_bundle}

FINAL INSTRUCTIONS:

1. Inspect every generated file listed by the deterministic report.

2. For every R# requirement, prove satisfied, unsatisfied, or uncertain
   using actual code evidence.

3. Trace the main modified execution path instead of only scanning for
   syntax.

4. Check cross-file calls, imports, routes, request/response shapes,
   selectors, templates, configuration keys, renamed symbols, function
   signatures, and shared state whenever relevant.

5. Use EXISTING FILE CHANGE CONTEXT to detect likely accidental
   regressions caused by complete file replacement.

6. Do not trust comments, TODOs, summaries, function names, or variable
   names as proof that behavior actually exists.

7. Do not report speculative defects. Blocking issues require concrete
   evidence.

8. reviewed_files must include EVERY generated file.

9. requirement_checks must include EXACTLY one entry for EVERY supplied
   R# item.

10. Return exactly the JSON object required by the system prompt.
'''

    raw = run_shared_generation(
        REVIEWER_SYSTEM,
        prompt,
        max_new_tokens=3600,
        temperature=0.0,
        do_sample=False,
    )

    normalized_review = (
        _review_normalize_result(
            raw,
            deterministic_report,
            requirements=requirements,
        )
    )

    del prompt
    del file_bundle
    del raw
    del existing_change_context

    clear_temporary_generation_memory()

    return normalized_review


def reviewer_step(
    plan: dict,
    code: str,
    force_rerun: bool = False,
    run_id: Optional[str] = None,
) -> str:
    """
    Step 4: strict semantic + deterministic review.

    Remains compatible with:
        /review
        /run-agent
        review.json

    Return format:
        PASS
        ...

    or:
        FAIL
        ...
    """

    if not force_rerun:

        cached = load_json(
            "review.json"
        )

        if cached:
            print(
                "📂 Using latest cached "
                "review.json record."
            )

            return str(
                cached.get("review")
                or ""
            )

    review = _run_reviewer_model(
        plan,
        code,
    )

    save_json(
        {
            "run_id": run_id,
            "project": _bob_context_project(),
            "review": review,
            "reviewer_version": (
                "semantic-deterministic-v3"
            ),
        },
        "review.json",
        meta={
            "stage": "reviewer",
            "run_id": run_id,
            "reviewer_version": (
                "semantic-deterministic-v3"
            ),
        },
    )

    print(
        "✅ Reviewer v3 appended "
        "to review.json"
    )

    return review


print(
    "✅ Reviewer accuracy patch v3 ready."
)
````

````python
# Phase 8 — Bob IDE compatible orchestrator
# MODE: BOTH — safe in single and multi mode

def _infer_final_status(review: str) -> str:
    return "PASS" if str(review or "").strip().upper().startswith("PASS") else "FAIL"


def _make_run_id(payload: Optional[dict] = None) -> str:
    existing = (payload or {}).get("run_id")
    if existing:
        return str(existing)
    return f"run_{uuid.uuid4().hex[:8]}"


def _extract_requested_file_contents(requested_files: List[str]) -> str:
    contents = ""
    for f in requested_files:
        rel = _normalize_path(f)
        c = read_file(rel)
        if c:
            contents += f"===== {rel} =====\n{c}\n\n"
        else:
            contents += f"===== {rel} NOT SUPPLIED =====\n\n"
    return contents


def plan_from_payload(payload: dict) -> dict:
    """HTTP /plan entrypoint. Returns a normalized plan object."""
    ctx = set_bob_payload_context(payload)
    run_id = _make_run_id(payload)
    user_prompt = payload.get("user_prompt") or payload.get("prompt") or payload.get("message") or ""
    workspace_tree = get_workspace_tree()
    files_payload = payload.get("files") or {}
    file_contents = _extract_requested_file_contents(list(files_payload.keys())) if files_payload else ""

    print(f"📥 /plan run_id={run_id} project={ctx['project']} files={ctx['file_count']}")
    plan = planner_step(user_prompt, workspace_tree, file_contents, force_rerun=True, run_id=run_id)
    return plan


def run_agent(user_prompt: str, max_iterations: int = 5, unload_model_at_end: bool = False,
              run_id: Optional[str] = None, return_proposals_only: bool = True) -> dict:
    """Run planner -> coder -> reviewer. Returns generated files as proposals; does not write project files."""
    run_id = run_id or f"run_{uuid.uuid4().hex[:8]}"
    workspace_tree = get_workspace_tree()
    file_contents = ""
    plan_data = None
    provided_files = set()
    workspace_request_count = 0
    file_changes = []
    files: Dict[str, Optional[str]] = {}

    _load_shared_model()

    for iteration in range(1, max_iterations + 1):
        print(f"\n--- Planning iteration {iteration} ---")
        plan_data = planner_step(user_prompt, workspace_tree, file_contents, force_rerun=True, run_id=run_id)
        mode = plan_data.get("output_mode")
        print(f"output_mode = {mode}")
        clear_temporary_generation_memory()

        if mode == "request_workspace":
            workspace_request_count += 1
            if workspace_tree or workspace_request_count >= 2:
                print("Workspace was already supplied. Forcing ready_for_coder.")
                plan_data["output_mode"] = "ready_for_coder"
                if not plan_data.get("coder_prompt"):
                    plan_data["coder_prompt"] = f"Create the necessary files for: {user_prompt}"
                break
            continue

        if mode == "request_files":
            requested_files = [_normalize_path(p) for p in plan_data.get("files_needed", [])]
            if not requested_files:
                plan_data["output_mode"] = "ready_for_coder"
                if not plan_data.get("coder_prompt"):
                    plan_data["coder_prompt"] = f"Make a best-effort implementation for: {user_prompt}"
                break

            new_files = [p for p in requested_files if p not in provided_files]
            if not new_files:
                print(f"Files {requested_files} already supplied. Forcing ready_for_coder.")
                plan_data["output_mode"] = "ready_for_coder"
                if not plan_data.get("coder_prompt"):
                    plan_data["coder_prompt"] = f"Modify the relevant files ({', '.join(requested_files)}) to satisfy the request."
                break

            file_contents += _extract_requested_file_contents(new_files)
            provided_files.update(new_files)
            # In Bob integration the backend sends capped files up front. Do not loop forever.
            if file_contents:
                continue
            plan_data["output_mode"] = "ready_for_coder"
            break

        if mode == "request_docs":
            print("Documentation requested. Not auto-supplied by Colab endpoint. Continuing best-effort.")
            plan_data["output_mode"] = "ready_for_coder"
            if not plan_data.get("coder_prompt"):
                plan_data["coder_prompt"] = f"Use supplied files only and make a best-effort implementation for: {user_prompt}"
            break

        if mode == "ready_for_coder":
            break

        print(f"Unknown mode {mode}. Forcing ready_for_coder.")
        plan_data["output_mode"] = "ready_for_coder"
        break
    else:
        print(f"Reached max iterations ({max_iterations}). Forcing ready_for_coder.")
        if plan_data is None:
            plan_data = normalize_plan({"summary": user_prompt, "output_mode": "ready_for_coder"})
        plan_data["output_mode"] = "ready_for_coder"
        if not plan_data.get("coder_prompt"):
            plan_data["coder_prompt"] = "Please implement the requested changes based on available information."

    print("Planner complete. Shared model remains loaded.")

    print("\n--- Building context ---")
    context = build_context_step(plan_data, force_rerun=True, run_id=run_id)

    print("--- Coding ---")
    code_out = coder_step(plan_data, context, force_rerun=True, run_id=run_id)
    clear_temporary_generation_memory()

    print("--- Reviewing ---")
    review = reviewer_step(plan_data, code_out, force_rerun=True, run_id=run_id)
    clear_temporary_generation_memory()

    # Retry only if reviewer failed and the model can improve output.
    for attempt in range(2):
        if _infer_final_status(review) == "PASS":
            break
        print(f"\nReview FAILED. Re-coding with feedback. Attempt {attempt + 1}/2.")
        plan_data["coder_prompt"] += f"\n\nReviewer feedback. Fix these issues:\n{review}"
        code_out = coder_step(plan_data, context, force_rerun=True, run_id=run_id)
        clear_temporary_generation_memory()
        review = reviewer_step(plan_data, code_out, force_rerun=True, run_id=run_id)
        clear_temporary_generation_memory()

    final_status = _infer_final_status(review)

    try:
        files = parse_coder_output(code_out, expected_files=plan_data.get("files_needed"))
    except Exception as e:
        print(f"⚠️ Could not parse code output: {e}")
        files = {}

    print("--- Building proposal change log. No files are written. ---")
    file_changes = save_files_with_change_log(files, run_id=run_id, review_status=final_status)

    change_record = create_change_record(run_id, user_prompt, plan_data, review, file_changes, final_status)
    save_json(change_record, CHANGELOG_FILE, meta={"stage": "orchestrator", "run_id": run_id})
    print(f"✅ Change log appended: {os.path.join(OUTPUT_DIR, CHANGELOG_FILE)}")

    response = {
        "run_id": run_id,
        "provider": "colab-qwen",
        "plan": plan_data,
        "code": code_out,
        "review": review,
        "final_status": final_status,
        "files": files,
        "metadata": {
            "model": SHARED_MODEL_NAME,
            "max_iterations": max_iterations,
            "project": _bob_context_project(),
            "active_path": _bob_context_active_path(),
            "proposal_count": len(file_changes),
            "colab_wrote_files": False,
            "change_log_path": os.path.join(OUTPUT_DIR, CHANGELOG_FILE),
            "run_log_path": os.path.join(OUTPUT_DIR, "runs.json"),
        },
        "proposals": file_changes,
    }

    run_record = {
        "run_id": run_id,
        "project": _bob_context_project(),
        "user_prompt": user_prompt,
        "final_status": final_status,
        "proposal_files": list(files.keys()),
        "change_log_file": CHANGELOG_FILE,
        "change_log_latest_index": len(load_json_history(CHANGELOG_FILE).get("records", [])) - 1,
    }
    save_json(run_record, "runs.json", meta={"stage": "orchestrator", "run_id": run_id})

    if unload_model_at_end:
        _unload_shared_model()
    else:
        print("Shared model kept loaded for next Bob IDE request.")

    return response


def run_agent_from_payload(payload: dict, max_iterations: int = 5) -> dict:
    """HTTP /run-agent entrypoint. Accepts Bob IDE payload and returns handoff contract response."""
    ctx = set_bob_payload_context(payload)
    run_id = _make_run_id(payload)
    user_prompt = payload.get("user_prompt") or payload.get("prompt") or payload.get("message") or ""
    print(f"📥 /run-agent run_id={run_id} project={ctx['project']} files={ctx['file_count']}")
    return run_agent(user_prompt, max_iterations=max_iterations, run_id=run_id, return_proposals_only=True)

print("✅ Bob IDE compatible orchestrator ready.")
````

````python
# Phase 8.5 — Lightning HTTP server for Bob IDE
# MODE: BOTH — safe in single and multi mode
# Exposes POST /plan and POST /run-agent exactly as expected by Bob IDE.
# In Colab, use ngrok or another tunnel and set BOB_COLAB_BASE_URL in your local Flask app.

_colab_http_app = None
_colab_http_thread = None
_colab_public_url = None


def _check_bearer_token(headers) -> Tuple[bool, str]:
    expected = os.environ.get("BOB_COLAB_TOKEN", "").strip()
    if not expected:
        return True, ""
    auth = headers.get("Authorization", "") if headers else ""
    if auth == f"Bearer {expected}":
        return True, ""
    return False, "Missing or invalid bearer token"


def create_colab_app():
    from flask import Flask, jsonify, request
    from flask_cors import CORS

    app = Flask(__name__)
    CORS(app)

    @app.get("/")
    @app.get("/health")
    def health():
        return jsonify({
            "ok": True,
            "service": "Bob Lightning Model Runtime",
            "model": SHARED_MODEL_NAME,
            "endpoints": ["/plan", "/run-agent"],
            "shared_model_loaded": _shared_model is not None,
        })

    @app.post("/plan")
    def http_plan():
        ok, error = _check_bearer_token(request.headers)
        if not ok:
            return jsonify({"error": error}), 401
        payload = request.get_json(force=True, silent=False) or {}
        try:
            plan = plan_from_payload(payload)
            # Bob accepts either the plan object directly or {"plan": plan}; return wrapper for clarity.
            return jsonify({"run_id": payload.get("run_id"), "provider": "colab-qwen", "plan": plan})
        except Exception as exc:
            traceback.print_exc()
            return jsonify({
                "run_id": payload.get("run_id"),
                "provider": "colab-qwen",
                "error": str(exc),
                "traceback": traceback.format_exc(),
            }), 500

    @app.post("/run-agent")
    def http_run_agent():
        ok, error = _check_bearer_token(request.headers)
        if not ok:
            return jsonify({"error": error}), 401
        payload = request.get_json(force=True, silent=False) or {}
        try:
            result = run_agent_from_payload(payload, max_iterations=int(payload.get("max_iterations", 5)))
            return jsonify(result)
        except Exception as exc:
            traceback.print_exc()
            return jsonify({
                "run_id": payload.get("run_id"),
                "provider": "colab-qwen",
                "final_status": "FAIL",
                "error": str(exc),
                "traceback": traceback.format_exc(),
                "files": {},
            }), 500

    return app


def start_colab_http_server(port: int = 8000, use_ngrok: bool = True, ngrok_auth_token: Optional[str] = None):
    """Start the Bob Lightning HTTP server.

    Usage in Colab:
      1. Set BOB_COLAB_TOKEN in Lightning/Colab environment secrets.
      2. start_colab_http_server(port=8000, use_ngrok=True)
      3. Copy the printed public URL into local backend env BOB_COLAB_BASE_URL.
    """
    global _colab_http_app, _colab_http_thread, _colab_public_url

    if use_ngrok and not os.environ.get("BOB_COLAB_TOKEN"):
        raise RuntimeError("BOB_COLAB_TOKEN is required before exposing the runtime publicly.")

    if _colab_http_thread and _colab_http_thread.is_alive():
        print("✅ Lightning HTTP server already running.")
        if _colab_public_url:
            print("Public URL:", _colab_public_url)
        return {"port": port, "public_url": _colab_public_url}

    _colab_http_app = create_colab_app()

    if use_ngrok:
        try:
            from pyngrok import ngrok
            token = ngrok_auth_token or os.environ.get("NGROK_AUTH_TOKEN")
            if token:
                ngrok.set_auth_token(token)
            tunnel = ngrok.connect(port, "http")
            _colab_public_url = str(tunnel.public_url)
            print("🌐 Public Bob Lightning URL:", _colab_public_url)
            print("Set locally:")
            print(f'$env:BOB_COLAB_BASE_URL="{_colab_public_url}"')
            print('$env:BOB_COLAB_PLAN_PATH="/plan"')
            print('$env:BOB_COLAB_RUN_PATH="/run-agent"')
            if os.environ.get("BOB_COLAB_TOKEN"):
                print('$env:BOB_COLAB_TOKEN="<same token you set in Colab>"')
        except Exception as exc:
            print("⚠️ Could not start ngrok tunnel:", exc)
            print(f"Server will still run locally on port {port}; expose it with your preferred tunnel.")

    def _run():
        _colab_http_app.run(host="0.0.0.0", port=port, debug=False, use_reloader=False, threaded=True)

    _colab_http_thread = threading.Thread(target=_run, daemon=True)
    _colab_http_thread.start()
    time.sleep(1)
    print(f"✅ Bob Lightning HTTP server running on port {port}")
    return {"port": port, "public_url": _colab_public_url}

print("✅ Lightning HTTP server helpers ready. Run start_colab_http_server() after loading/authenticating the model.")
````

````python
# Phase 13 - Bob IDE Colab v2 HTTP contract
# MODE: BOTH — safe in single and multi mode
# This overrides create_colab_app() with the endpoints negotiated by Bob Chat.

BOB_COLAB_CONTRACT_VERSION = "bob-colab-v2"
_bob_colab_runs: Dict[str, dict] = {}
_bob_colab_cancelled = set()


def _bob_run_event(run_id: str, status: str, **extra):
    record = {
        "run_id": run_id,
        "status": status,
        "updated_at": now_iso(),
        **extra,
    }
    _bob_colab_runs[run_id] = {**_bob_colab_runs.get(run_id, {}), **record}
    return record


def create_colab_app():
    from flask import Flask, Response, jsonify, request, stream_with_context
    from flask_cors import CORS

    app = Flask(__name__)
    CORS(app)

    def authorized():
        ok, error = _check_bearer_token(request.headers)
        if not ok:
            return None, (jsonify({"error": error}), 401)
        return request.get_json(force=True, silent=True) or {}, None

    @app.get("/")
    @app.get("/health")
    def health():
        return jsonify({
            "ok": True,
            "service": "Bob Lightning Model Runtime",
            "contract_version": BOB_COLAB_CONTRACT_VERSION,
            "model": SHARED_MODEL_NAME,
            "gpu_available": bool(torch.cuda.is_available()),
            "shared_model_loaded": _shared_model is not None,
            "streaming": True,
        })

    @app.get("/capabilities")
    def capabilities():
        return jsonify({
            "contract_version": BOB_COLAB_CONTRACT_VERSION,
            "model": SHARED_MODEL_NAME,
            "modes": ["chat", "plan", "agent"],
            "streaming": True,
            "cancellation": True,
            "max_iterations": {"minimum": 1, "maximum": 20, "default": 5},
            "context_modes": ["active", "open", "workspace"],
            "endpoints": {
                "health": "/health",
                "capabilities": "/capabilities",
                "chat": "/chat",
                "plan": "/plan",
                "run": "/run-agent",
                "stream": "/run-agent/stream",
                "status": "/runs/{run_id}",
                "cancel": "/runs/{run_id}/cancel",
            },
        })

    @app.post("/chat")
    def chat():
        payload, error = authorized()
        if error:
            return error
        set_bob_payload_context(payload)
        prompt = payload.get("user_prompt") or payload.get("message") or ""
        system = (
            "You are Bob, a concise coding assistant. Answer questions using the supplied "
            "workspace context. Do not claim to have changed files and do not emit proposals."
        )
        context = get_workspace_tree()
        active = read_file(_bob_context_active_path()) if _bob_context_active_path() else ""
        reply = run_shared_generation(
            system,
            f"Workspace:\n{context}\n\nActive file:\n{active[:12000]}\n\nUser:\n{prompt}",
            max_new_tokens=1200,
        )
        return jsonify({"provider": "colab-qwen", "reply": reply, "run_id": payload.get("run_id")})

    @app.post("/plan")
    def http_plan():
        payload, error = authorized()
        if error:
            return error
        try:
            plan = plan_from_payload(payload)
            return jsonify({"run_id": payload.get("run_id"), "provider": "colab-qwen", "plan": plan})
        except Exception as exc:
            traceback.print_exc()
            return jsonify({"run_id": payload.get("run_id"), "error": str(exc)}), 500

    def execute_agent(payload):
        run_id = payload.get("run_id") or _make_run_id(payload)
        _bob_run_event(run_id, "running")
        if run_id in _bob_colab_cancelled:
            return _bob_run_event(run_id, "cancelled")
        try:
            result = run_agent_from_payload(payload, max_iterations=int(payload.get("max_iterations", 5)))
            if run_id in _bob_colab_cancelled:
                return _bob_run_event(run_id, "cancelled")
            _bob_run_event(run_id, "completed", result=result)
            return result
        except Exception as exc:
            traceback.print_exc()
            _bob_run_event(run_id, "failed", error=str(exc))
            raise

    @app.post("/run-agent")
    def http_run_agent():
        payload, error = authorized()
        if error:
            return error
        try:
            return jsonify(execute_agent(payload))
        except Exception as exc:
            return jsonify({
                "run_id": payload.get("run_id"),
                "provider": "colab-qwen",
                "final_status": "FAIL",
                "error": str(exc),
                "files": {},
            }), 500

    @app.post("/run-agent/stream")
    def http_run_agent_stream():
        payload, error = authorized()
        if error:
            return error
        run_id = payload.get("run_id") or _make_run_id(payload)
        payload["run_id"] = run_id

        @stream_with_context
        def generate():
            yield f"data: {json.dumps(_bob_run_event(run_id, 'queued'))}\n\n"
            yield f"data: {json.dumps(_bob_run_event(run_id, 'planning'))}\n\n"
            try:
                result = execute_agent(payload)
                status = "cancelled" if result.get("status") == "cancelled" else "completed"
                yield f"data: {json.dumps({'run_id': run_id, 'status': status, 'result': result})}\n\n"
            except Exception as exc:
                yield f"data: {json.dumps({'run_id': run_id, 'status': 'failed', 'error': str(exc)})}\n\n"

        return Response(generate(), mimetype="text/event-stream")

    @app.get("/runs/<run_id>")
    def run_status(run_id):
        return jsonify(_bob_colab_runs.get(run_id, {"run_id": run_id, "status": "unknown"}))

    @app.post("/runs/<run_id>/cancel")
    def cancel_run(run_id):
        _bob_colab_cancelled.add(run_id)
        return jsonify(_bob_run_event(run_id, "cancelling"))

    return app


print("Bob Colab v2 contract ready:", BOB_COLAB_CONTRACT_VERSION)
print("Run start_colab_http_server() to expose it.")
````

````python
# MODE: BOTH — uses the Phase 0.2A selection

# Phase 14 — Staged Bob Chat Colab contract (Plan → Code → Review → Run All)
# This phase intentionally overrides the older create_colab_app() so Bob IDE can
# drive each model stage independently and keep every generated file as a proposal.

BOB_COLAB_CONTRACT_VERSION = "bob-colab-v4-llmops"


# BOB_STRUCTURED_RUNTIME_LOGGING_V1_BEGIN
# Persistent JSONL observability for Lightning AI/Colab. Raw prompts, generated
# code, workspace contents, credentials, and request headers are never logged.
import hashlib as _bob_hashlib
import threading as _bob_log_threading
import urllib.parse as _bob_urlparse
import uuid as _bob_uuid
from datetime import datetime as _bob_datetime, timezone as _bob_timezone
from pathlib import Path as _BobPath

BOB_RUNTIME_LOG_DIR = _BobPath(os.environ.get("BOB_RUNTIME_LOG_DIR", "bob_runtime_logs")).resolve()
BOB_RUNTIME_LOG_DIR.mkdir(parents=True, exist_ok=True)
_BOB_RUNTIME_LOG_FILES = {
    "app": BOB_RUNTIME_LOG_DIR / "app-events.jsonl",
    "model": BOB_RUNTIME_LOG_DIR / "model-events.jsonl",
    "ngrok": BOB_RUNTIME_LOG_DIR / "ngrok-events.jsonl",
}
_BOB_RUNTIME_LOG_LOCK = globals().get("_BOB_RUNTIME_LOG_LOCK", _bob_log_threading.RLock())
_BOB_RUNTIME_LOG_MAX_BYTES = max(1024 * 1024, int(os.environ.get("BOB_RUNTIME_LOG_MAX_BYTES", str(10 * 1024 * 1024))))
_BOB_LOG_DROPPED_KEYS = {
    "authorization", "cookie", "set-cookie", "password", "password_hash",
    "session", "session_token", "token", "token_hash", "secret", "api_key",
    "code", "content", "files", "generated_code", "output", "request_body",
    "response_body", "headers",
}


def _bob_log_timestamp() -> str:
    return _bob_datetime.now(_bob_timezone.utc).isoformat().replace("+00:00", "Z")


def _bob_configured_secrets():
    values = [
        os.environ.get("BOB_COLAB_TOKEN", ""),
        os.environ.get("HF_TOKEN", ""),
        os.environ.get("NGROK_AUTH_TOKEN", ""),
        os.environ.get("NGROK_AUTHTOKEN", ""),
    ]
    values.extend(os.environ.get("BOB_REDACT_VALUES", "").split(","))
    return [value for value in values if value]


def _bob_redact_log_text(value) -> str:
    text = str(value or "")
    text = re.sub(r"(?i)Bearer\s+[A-Za-z0-9._~+/=-]+", "Bearer [REDACTED]", text)
    text = re.sub(r"(?i)\b(sk-|ghp_|github_pat_|hf_|ngrok_)[A-Za-z0-9_-]{8,}\b", "[REDACTED_TOKEN]", text)
    text = re.sub(r"(?i)(password|passwd|api[_-]?key|secret|token)\s*[:=]\s*[^\s,;]+", r"\1=[REDACTED]", text)
    for secret in _bob_configured_secrets():
        text = text.replace(secret, "[REDACTED_CONFIGURED_SECRET]")
    return text[:100000]


def _bob_sanitize_log_value(value, key=""):
    if str(key).lower() in _BOB_LOG_DROPPED_KEYS:
        return None
    if isinstance(value, str):
        return _bob_redact_log_text(value)
    if isinstance(value, (int, float, bool)) or value is None:
        return value
    if isinstance(value, (list, tuple)):
        return [_bob_sanitize_log_value(item) for item in value]
    if isinstance(value, dict):
        return {
            str(name): safe
            for name, item in value.items()
            if (safe := _bob_sanitize_log_value(item, name)) is not None
        }
    return _bob_redact_log_text(value)


def _bob_prompt_metadata(payload: dict) -> dict:
    prompt = str((payload or {}).get("user_prompt") or (payload or {}).get("prompt") or (payload or {}).get("message") or "")
    encoded = prompt.encode("utf-8")
    return {"prompt_sha256": _bob_hashlib.sha256(encoded).hexdigest(), "prompt_size_bytes": len(encoded)} if prompt else {}


def _bob_runtime_log(source: str, event: str, **payload) -> None:
    file_path = _BOB_RUNTIME_LOG_FILES[source]
    record = _bob_sanitize_log_value({"timestamp": _bob_log_timestamp(), "event": event, **payload})
    line = json.dumps(record, ensure_ascii=False, separators=(",", ":")) + "\n"
    with _BOB_RUNTIME_LOG_LOCK:
        try:
            if file_path.exists() and file_path.stat().st_size >= _BOB_RUNTIME_LOG_MAX_BYTES:
                backup = file_path.with_suffix(file_path.suffix + ".1")
                if backup.exists():
                    backup.unlink()
                file_path.replace(backup)
            with file_path.open("a", encoding="utf-8") as handle:
                handle.write(line)
                handle.flush()
        except Exception as log_error:
            print("Structured runtime log error:", _bob_redact_log_text(log_error))


print("Structured runtime logs:", BOB_RUNTIME_LOG_DIR)
# BOB_STRUCTURED_RUNTIME_LOGGING_V1_END
_bob_colab_runs = globals().get("_bob_colab_runs", {})
_bob_colab_cancelled = globals().get("_bob_colab_cancelled", set())
_bob_request_lock = globals().get("_bob_request_lock", threading.RLock())  # legacy compatibility only

# BOB_CONFIGURABLE_MODEL_LANES_V1_BEGIN
from contextlib import contextmanager as _bob_contextmanager

_bob_pipeline_state_lock = globals().get("_bob_pipeline_state_lock", threading.RLock())
_bob_lane_locks = [threading.RLock() for _ in range(BOB_MODEL_LANE_COUNT)]
_bob_lane_owners = [None for _ in range(BOB_MODEL_LANE_COUNT)]
_bob_pipeline_assignments = {}
_bob_pipeline_stages = {}
_bob_pipeline_last_seen = {}
_BOB_PIPELINE_TTL_SECONDS = max(300, int(os.environ.get("BOB_PIPELINE_TTL_SECONDS", "3600")))


class BobPipelineLaneError(RuntimeError):
    pass


def _pipeline_id(payload: dict) -> str:
    value = payload.get("pipeline_id") or payload.get("run_id") or payload.get("trace_id")
    if not value:
        raise BobPipelineLaneError("pipeline_id or run_id is required for model execution")
    return str(value)


def _requested_lane(payload: dict):
    value = payload.get("model_lane")
    if value is None or value == "":
        return None
    try:
        lane = int(value)
    except (TypeError, ValueError) as exc:
        raise BobPipelineLaneError("model_lane must be an integer") from exc
    if lane < 0 or lane >= BOB_MODEL_LANE_COUNT:
        raise BobPipelineLaneError(f"model_lane must be between 0 and {BOB_MODEL_LANE_COUNT - 1}")
    return lane


def _prune_stale_pipelines(timestamp: float) -> None:
    stale = [pipeline_id for pipeline_id, seen in _bob_pipeline_last_seen.items() if timestamp - seen > _BOB_PIPELINE_TTL_SECONDS]
    for pipeline_id in stale:
        lane = _bob_pipeline_assignments.pop(pipeline_id, None)
        _bob_pipeline_stages.pop(pipeline_id, None)
        _bob_pipeline_last_seen.pop(pipeline_id, None)
        if lane is not None and _bob_lane_owners[lane] == pipeline_id:
            _bob_lane_owners[lane] = None


def _reserve_pipeline(payload: dict) -> tuple[str, int]:
    pipeline_id = _pipeline_id(payload)
    requested = _requested_lane(payload)
    timestamp = time.monotonic()
    with _bob_pipeline_state_lock:
        _prune_stale_pipelines(timestamp)
        assigned = _bob_pipeline_assignments.get(pipeline_id)
        if assigned is not None:
            if requested is not None and requested != assigned:
                raise BobPipelineLaneError(
                    f"Pipeline {pipeline_id} is already bound to lane {assigned}; lane switching is forbidden"
                )
            _bob_pipeline_last_seen[pipeline_id] = timestamp
            return pipeline_id, assigned
        candidate_lanes = [requested] if requested is not None else list(range(BOB_MODEL_LANE_COUNT))
        lane = next((item for item in candidate_lanes if _bob_lane_owners[item] is None), None)
        if lane is None:
            raise BobPipelineLaneError("All model lanes are busy; retry with the same pipeline_id")
        _bob_pipeline_assignments[pipeline_id] = lane
        _bob_lane_owners[lane] = pipeline_id
        _bob_pipeline_last_seen[pipeline_id] = timestamp
        return pipeline_id, lane


def _validate_pipeline_stage(pipeline_id: str, stage: str) -> None:
    previous = _bob_pipeline_stages.get(pipeline_id)
    allowed = {
        None: {"chat", "plan", "code", "run-agent"},
        "plan": {"plan", "replan", "code"},
        "replan": {"replan", "code"},
        "code": {"code", "review"},
        "chat": set(),
        "review": set(),
        "run-agent": set(),
    }
    if stage not in allowed.get(previous, set()):
        raise BobPipelineLaneError(f"Invalid pipeline stage transition {previous!r} -> {stage!r}")
    _bob_pipeline_stages[pipeline_id] = stage


def _release_pipeline(pipeline_id: str, lane: int) -> None:
    with _bob_pipeline_state_lock:
        if _bob_pipeline_assignments.get(pipeline_id) == lane:
            _bob_pipeline_assignments.pop(pipeline_id, None)
            _bob_pipeline_stages.pop(pipeline_id, None)
            _bob_pipeline_last_seen.pop(pipeline_id, None)
            if _bob_lane_owners[lane] == pipeline_id:
                _bob_lane_owners[lane] = None


@_bob_contextmanager
def _bob_pipeline_guard(payload: dict, stage: str, release: bool = False):
    pipeline_id, lane = _reserve_pipeline(payload)
    failed = False
    with _bob_lane_locks[lane]:
        with _bob_pipeline_state_lock:
            if _bob_lane_owners[lane] != pipeline_id:
                raise BobPipelineLaneError("Pipeline lane ownership changed unexpectedly")
            _validate_pipeline_stage(pipeline_id, stage)
            _bob_pipeline_last_seen[pipeline_id] = time.monotonic()
        previous_lane = getattr(_bob_runtime_state, "model_lane", None)
        previous_pipeline = getattr(_bob_runtime_state, "pipeline_id", None)
        _bob_runtime_state.model_lane = lane
        _bob_runtime_state.pipeline_id = pipeline_id
        payload["pipeline_id"] = pipeline_id
        payload["model_lane"] = lane
        try:
            yield lane
        except Exception:
            failed = True
            raise
        finally:
            clear_bob_payload_context()
            if previous_lane is None:
                if hasattr(_bob_runtime_state, "model_lane"):
                    delattr(_bob_runtime_state, "model_lane")
            else:
                _bob_runtime_state.model_lane = previous_lane
            if previous_pipeline is None:
                if hasattr(_bob_runtime_state, "pipeline_id"):
                    delattr(_bob_runtime_state, "pipeline_id")
            else:
                _bob_runtime_state.pipeline_id = previous_pipeline
            if release or failed:
                _release_pipeline(pipeline_id, lane)


def _lane_health() -> dict:
    with _bob_pipeline_state_lock:
        return {
            "runtime_mode": BOB_RUNTIME_MODE,
            "model_lane_count": BOB_MODEL_LANE_COUNT,
            "configured_multi_lane_count": BOB_MULTI_LANE_COUNT,
            "loaded_model_lanes": [index for index, model in enumerate(_bob_model_lanes) if model is not None],
            "active_pipeline_count": len(_bob_pipeline_assignments),
            "sticky_pipeline_routing": BOB_RUNTIME_MODE == "multi",
            "single_lane_serialization": BOB_RUNTIME_MODE == "single",
            "pipeline_stage_validation": True,
            "request_context_isolation": "thread_local",
        }
# BOB_CONFIGURABLE_MODEL_LANES_V1_END


def _current_usage() -> dict:
    usage = getattr(_bob_runtime_state, "usage", None) or {}
    return {
        "input_tokens": max(0, int(usage.get("input_tokens") or 0)),
        "output_tokens": max(0, int(usage.get("output_tokens") or 0)),
        "total_tokens": max(0, int(usage.get("total_tokens") or 0)),
    }


def _sum_usage(*items: dict) -> dict:
    return {
        key: sum(int((item or {}).get(key) or 0) for item in items)
        for key in ("input_tokens", "output_tokens", "total_tokens")
    }


def _safe_error_message(exc: Exception) -> str:
    text = str(exc or "Runtime error")
    for secret in (os.environ.get("BOB_COLAB_TOKEN"), os.environ.get("HF_TOKEN"), os.environ.get("NGROK_AUTH_TOKEN")):
        if secret and len(secret) >= 6:
            text = text.replace(secret, "[REDACTED]")
    text = re.sub(r"(?i)Bearer\s+[A-Za-z0-9._~+/=-]+", "Bearer [REDACTED]", text)
    return text[:1000]


def _request_metadata(payload: dict) -> dict:
    run_id = _payload_run_id(payload)
    return {
        "run_id": run_id,
        "trace_id": payload.get("trace_id") or run_id,
        "request_id": payload.get("request_id"),
        "actor_user_id": payload.get("actor_user_id"),
        "evaluation_run_id": payload.get("evaluation_run_id"),
        "test_id": payload.get("test_id"),
        "test_name": payload.get("test_name"),
        "prompt_category": payload.get("prompt_category"),
        "pair_id": payload.get("pair_id"),
        "approach": payload.get("approach"),
        "pipeline_id": payload.get("pipeline_id") or payload.get("run_id"),
        "model_lane": payload.get("model_lane", getattr(_bob_runtime_state, "model_lane", None)),
        "model": SHARED_MODEL_NAME,
        "model_revision": os.environ.get("BOB_MODEL_REVISION", "unversioned"),
        "prompt_version": os.environ.get("BOB_PROMPT_SET_VERSION", "notebook-v4"),
        "contract_version": BOB_COLAB_CONTRACT_VERSION,
    }


def _response_metadata(payload: dict, stage: str, started: float, usage: Optional[dict] = None) -> dict:
    from datetime import datetime, timezone, timedelta
    duration = max(0.0, time.perf_counter() - started)
    finished = datetime.now(timezone.utc)
    metadata = {
        **_request_metadata(payload),
        "provider": "colab-qwen",
        "stage": stage,
        "started_at": (finished - timedelta(seconds=duration)).isoformat().replace("+00:00", "Z"),
        "finished_at": finished.isoformat().replace("+00:00", "Z"),
        "duration_ms": round(duration * 1000),
        "attempt_count": 1,
        "usage": usage or _current_usage(),
    }
    _bob_runtime_log("model", "model.stage", **metadata)
    print(json.dumps({"event": "model.stage", **metadata}, ensure_ascii=False))
    return metadata


def _structured_error(payload: dict, stage: str, exc: Exception, started: float) -> dict:
    error = {
        "type": type(exc).__name__,
        "component": "lightning_runtime",
        "stage": stage,
        "message": _safe_error_message(exc),
        "retriable": False,
    }
    metadata = _response_metadata(payload, stage, started, {"input_tokens": 0, "output_tokens": 0, "total_tokens": 0})
    _bob_runtime_log("model", "model.error", **metadata, error=error)
    print(json.dumps({"event": "model.error", **metadata, "error": error}, ensure_ascii=False))
    return {**metadata, "error": error}


def _check_bearer_token(headers) -> Tuple[bool, str]:
    import hmac
    expected = os.environ.get("BOB_COLAB_TOKEN", "")
    if not expected:
        return False, "BOB_COLAB_TOKEN is not configured"
    supplied = headers.get("Authorization", "") if headers else ""
    valid = hmac.compare_digest(str(supplied), f"Bearer {expected}")
    return (True, "") if valid else (False, "Missing or invalid bearer token")


def _payload_prompt(payload: dict) -> str:
    return payload.get("user_prompt") or payload.get("prompt") or payload.get("message") or ""


def _payload_run_id(payload: dict) -> str:
    return payload.get("run_id") or _make_run_id(payload)


def _set_staged_payload_context(payload: dict) -> dict:
    """Merge app-selected files and forced files into the active notebook context."""
    payload = dict(payload or {})
    files = dict(payload.get("files") or {})
    forced = dict(payload.get("forced_files") or {})
    # forced files intentionally win if duplicated
    merged = {**files, **forced}
    payload["files"] = merged
    payload["forced_files"] = forced
    return set_bob_payload_context(payload)


def _context_used(payload: dict) -> dict:
    files = dict(payload.get("files") or {})
    forced = set((payload.get("forced_files") or {}).keys())
    records = []
    for file_path, content in files.items():
        encoded = str(content or "").encode("utf-8")
        records.append({
            "path": _normalize_path(file_path),
            "sha256": hashlib.sha256(encoded).hexdigest(),
            "size_bytes": len(encoded),
            "selection_reason": "forced" if file_path in forced else "context_selected",
        })
    return {
        "files": records,
        "total_bytes": sum(item["size_bytes"] for item in records),
        "context_policy": payload.get("context_policy") or {},
    }


def plan_from_payload(payload: dict) -> dict:
    """POST /plan: plan only; no code; no file writes."""
    ctx = _set_staged_payload_context(payload)
    run_id = _payload_run_id(payload)
    prompt = _payload_prompt(payload)
    workspace_tree = get_workspace_tree()
    file_contents = _extract_requested_file_contents(list((payload.get("files") or {}).keys()))
    print(f"📥 /plan run_id={run_id} project={ctx['project']} files={ctx['file_count']} forced={len(payload.get('forced_files') or {})}")
    plan = planner_step(prompt, workspace_tree, file_contents, force_rerun=True, run_id=run_id)
    return plan


def replan_from_payload(payload: dict) -> dict:
    """POST /replan: plan again using the selected/previous plan plus forced files."""
    ctx = _set_staged_payload_context(payload)
    run_id = _payload_run_id(payload)
    prompt = _payload_prompt(payload)
    previous_plan = payload.get("previous_plan") or payload.get("selected_plan") or payload.get("plan") or {}
    file_contents = _extract_requested_file_contents(list((payload.get("files") or {}).keys()))
    workspace_tree = get_workspace_tree()
    if previous_plan:
        file_contents += "\n===== PREVIOUS SELECTED PLAN =====\n" + json.dumps(previous_plan, indent=2, ensure_ascii=False) + "\n"
    print(f"📥 /replan run_id={run_id} previous={payload.get('previous_plan_id')} forced={len(payload.get('forced_files') or {})}")
    plan = planner_step(prompt, workspace_tree, file_contents, force_rerun=True, run_id=run_id)
    return plan


def code_from_payload(payload: dict) -> dict:
    """POST /code: code selected plan only; return full generated file contents."""
    _set_staged_payload_context(payload)
    run_id = _payload_run_id(payload)
    plan = normalize_plan(payload.get("selected_plan") or payload.get("plan") or {})
    if not plan.get("coder_prompt"):
        plan["coder_prompt"] = _payload_prompt(payload) or plan.get("summary") or "Implement the selected plan."
    context = build_context_step(plan, force_rerun=True, run_id=run_id)
    started = time.perf_counter()
    code = coder_step(plan, context, force_rerun=True, run_id=run_id)
    files = parse_coder_output(code, expected_files=plan.get("files_needed"))
    return {
        "run_id": run_id,
        "plan_id": payload.get("plan_id"),
        "provider": "colab-qwen",
        "stage": "coder",
        "plan": plan,
        "code": code,
        "files": files,
        "context_used": _context_used(payload),
        **_response_metadata(payload, "coder", started),
    }


def review_from_payload(payload: dict) -> dict:
    """POST /review: review coder output. Does not write files."""
    _set_staged_payload_context(payload)
    run_id = _payload_run_id(payload)
    plan = normalize_plan(payload.get("selected_plan") or payload.get("plan") or {})
    code = payload.get("code") or ""
    files = payload.get("files") or {}
    if not code and isinstance(files, dict):
        code = "\n\n".join(f"### `{path}`\n```\n{content or ''}\n```" for path, content in files.items())
    started = time.perf_counter()
    review = reviewer_step(plan, code, force_rerun=True, run_id=run_id)
    status = "PASS" if str(review).lstrip().upper().startswith("PASS") else "FAIL"
    return {
        "run_id": run_id,
        "plan_id": payload.get("plan_id"),
        "provider": "colab-qwen",
        "stage": "reviewer",
        "review": review,
        "final_status": status,
        "files": files,
        **_response_metadata(payload, "reviewer", started),
    }


def run_agent_from_payload(payload: dict, max_iterations: int = 5) -> dict:
    """POST /run-agent shortcut: plan → code → review; returns proposal-ready files only."""
    run_id = _payload_run_id(payload)
    prompt = _payload_prompt(payload)
    started = time.perf_counter()
    plan = plan_from_payload(payload)
    plan_usage = _current_usage()
    code_result = code_from_payload({**payload, "run_id": run_id, "plan": plan, "selected_plan": plan})
    review_result = review_from_payload({**payload, "run_id": run_id, "plan": plan, "selected_plan": plan, "code": code_result["code"], "files": code_result["files"]})
    total_usage = _sum_usage(plan_usage, code_result.get("usage"), review_result.get("usage"))
    return {
        "run_id": run_id,
        "provider": "colab-qwen",
        "plan": plan,
        "code": code_result["code"],
        "review": review_result["review"],
        "final_status": review_result["final_status"],
        "files": code_result["files"],
        **_response_metadata(payload, "agent", started, total_usage),
        "metadata": {
            "model": SHARED_MODEL_NAME,
            "model_revision": os.environ.get("BOB_MODEL_REVISION", "unversioned"),
            "prompt_version": os.environ.get("BOB_PROMPT_SET_VERSION", "notebook-v4"),
            "contract_version": BOB_COLAB_CONTRACT_VERSION,
            "context_used": code_result.get("context_used", {}),
            "max_iterations": max_iterations,
        },
    }


def create_colab_app():
    from flask import Flask, Response, g, jsonify, request, stream_with_context

    app = Flask(__name__)
    app.config["MAX_CONTENT_LENGTH"] = int(os.environ.get("BOB_MAX_REQUEST_BYTES", str(50 * 1024 * 1024)))

    @app.before_request
    def structured_request_start():
        g.bob_request_started = time.perf_counter()
        payload = request.get_json(silent=True) or {}
        run_id = payload.get("run_id")
        g.bob_request_id = payload.get("request_id") or request.headers.get("X-Request-ID") or f"req_{_bob_uuid.uuid4().hex}"
        g.bob_trace_id = payload.get("trace_id") or run_id
        g.bob_run_id = run_id
        g.bob_actor_user_id = payload.get("actor_user_id")
        g.bob_evaluation_metadata = {name: payload.get(name) for name in ("evaluation_run_id", "test_id", "test_name", "prompt_category", "pair_id", "approach", "pipeline_id", "model_lane")}
        g.bob_prompt_metadata = _bob_prompt_metadata(payload)

    @app.after_request
    def structured_request_finish(response):
        duration_ms = round(max(0.0, time.perf_counter() - getattr(g, "bob_request_started", time.perf_counter())) * 1000)
        metadata = {
            "request_id": getattr(g, "bob_request_id", None),
            "trace_id": getattr(g, "bob_trace_id", None),
            "run_id": getattr(g, "bob_run_id", None),
            "actor_user_id": getattr(g, "bob_actor_user_id", None),
            "method": request.method,
            "path": request.path,
            "status": response.status_code,
            "duration_ms": duration_ms,
            "request_size_bytes": request.content_length or 0,
            "response_size_bytes": response.calculate_content_length(),
            "outcome": "success" if response.status_code < 400 else "error",
            **getattr(g, "bob_evaluation_metadata", {}),
            **getattr(g, "bob_prompt_metadata", {}),
        }
        _bob_runtime_log("app", "app.http_request", **metadata)
        forwarded_host = request.headers.get("X-Forwarded-Host", "")
        host = forwarded_host or request.host
        if "ngrok" in host.lower():
            _bob_runtime_log("ngrok", "ngrok.http_request", host=host, forwarded_proto=request.headers.get("X-Forwarded-Proto"), **metadata)
        response.headers["X-Request-ID"] = metadata["request_id"]
        return response

    def authorized_json():
        ok, error = _check_bearer_token(request.headers)
        if not ok:
            return None, (jsonify({"error": error}), 401)
        return request.get_json(force=True, silent=True) or {}, None

    @app.get("/")
    @app.get("/health")
    def health():
        return jsonify({
            "ok": True,
            "service": "Bob Lightning Model Runtime",
            "contract_version": BOB_COLAB_CONTRACT_VERSION,
            "model": SHARED_MODEL_NAME,
            "gpu_available": bool(torch.cuda.is_available()),
            "shared_model_loaded": any(model is not None for model in _bob_model_lanes),
            "streaming": True,
            "authentication_required": True,
            "usage_reporting": True,
            "request_tracing": True,
            "structured_logging": True,
            "runtime_log_dir": str(BOB_RUNTIME_LOG_DIR),
            "concurrency": _lane_health(),
            "endpoints": ["/chat", "/plan", "/replan", "/code", "/review", "/run-agent", "/run-agent/stream", "/capabilities"],
        })

    @app.get("/capabilities")
    def capabilities():
        return jsonify({
            "contract_version": BOB_COLAB_CONTRACT_VERSION,
            "model": SHARED_MODEL_NAME,
            "modes": ["chat", "plan", "replan", "code", "review", "agent"],
            "streaming": True,
            "cancellation": True,
            "authentication_required": True,
            "usage_reporting": True,
            "request_tracing": True,
            "structured_logging": True,
            "runtime_log_dir": str(BOB_RUNTIME_LOG_DIR),
            "concurrency": _lane_health(),
            "endpoints": {
                "health": "/health",
                "capabilities": "/capabilities",
                "chat": "/chat",
                "plan": "/plan",
                "replan": "/replan",
                "code": "/code",
                "review": "/review",
                "run": "/run-agent",
                "stream": "/run-agent/stream",
                "status": "/runs/{run_id}",
                "cancel": "/runs/{run_id}/cancel",
            },
        })

    @app.post("/chat")
    def chat():
        payload, error = authorized_json()
        if error: return error
        prompt = _payload_prompt(payload)
        system = "You are Bob, a concise coding assistant. Use supplied workspace context. Do not claim to have changed files."
        started = time.perf_counter()
        with _bob_pipeline_guard(payload, "chat", release=True):
            _set_staged_payload_context(payload)
            reply = run_shared_generation(system, f"Workspace:\n{get_workspace_tree()}\n\nUser:\n{prompt}", max_new_tokens=1200)
        return jsonify({"reply": reply, **_response_metadata(payload, "chat", started)})

    @app.post("/plan")
    def http_plan():
        payload, error = authorized_json()
        if error: return error
        started = time.perf_counter()
        try:
            with _bob_pipeline_guard(payload, "plan", release=bool(payload.get("release_pipeline"))):
                plan = plan_from_payload(payload)
            return jsonify({"plan": plan, "context_used": _context_used(payload), **_response_metadata(payload, "planner", started)})
        except Exception as exc:
            print("Planner error:", _safe_error_message(exc))
            return jsonify(_structured_error(payload, "planner", exc, started)), 500

    @app.post("/replan")
    def http_replan():
        payload, error = authorized_json()
        if error: return error
        started = time.perf_counter()
        try:
            with _bob_pipeline_guard(payload, "replan", release=bool(payload.get("release_pipeline"))):
                plan = replan_from_payload(payload)
            return jsonify({"previous_plan_id": payload.get("previous_plan_id"), "plan": plan, "context_used": _context_used(payload), **_response_metadata(payload, "planner", started)})
        except Exception as exc:
            print("Replan error:", _safe_error_message(exc))
            return jsonify(_structured_error(payload, "planner", exc, started)), 500

    @app.post("/code")
    def http_code():
        payload, error = authorized_json()
        if error: return error
        started = time.perf_counter()
        try:
            with _bob_pipeline_guard(payload, "code", release=bool(payload.get("release_pipeline"))):
                return jsonify(code_from_payload(payload))
        except Exception as exc:
            print("Coder error:", _safe_error_message(exc))
            return jsonify({**_structured_error(payload, "coder", exc, started), "files": {}}), 500

    @app.post("/review")
    def http_review():
        payload, error = authorized_json()
        if error: return error
        started = time.perf_counter()
        try:
            with _bob_pipeline_guard(payload, "review", release=True):
                return jsonify(review_from_payload(payload))
        except Exception as exc:
            print("Reviewer error:", _safe_error_message(exc))
            safe_error = _safe_error_message(exc)
            return jsonify({**_structured_error(payload, "reviewer", exc, started), "final_status": "FAIL", "review": f"FAIL\n{safe_error}"}), 500

    @app.post("/run-agent")
    def http_run_agent():
        payload, error = authorized_json()
        if error: return error
        started = time.perf_counter()
        try:
            with _bob_pipeline_guard(payload, "run-agent", release=True):
                return jsonify(run_agent_from_payload(payload, max_iterations=int(payload.get("max_iterations", 5))))
        except Exception as exc:
            print("Agent error:", _safe_error_message(exc))
            return jsonify({**_structured_error(payload, "agent", exc, started), "final_status": "FAIL", "files": {}}), 500

    @app.post("/run-agent/stream")
    def http_stream():
        payload, error = authorized_json()
        if error: return error
        def generate():
            run_id = _payload_run_id(payload)
            for status in ["planning", "coding", "reviewing"]:
                yield "data: " + json.dumps({"run_id": run_id, "status": status}) + "\n\n"
            with _bob_pipeline_guard(payload, "run-agent", release=True):
                result = run_agent_from_payload(payload, max_iterations=int(payload.get("max_iterations", 5)))
            yield "data: " + json.dumps({"run_id": run_id, "status": "completed", "result": result}) + "\n\n"
        return Response(stream_with_context(generate()), mimetype="text/event-stream")

    @app.get("/runs/<run_id>")
    def run_status(run_id):
        ok, error = _check_bearer_token(request.headers)
        if not ok: return jsonify({"error": error}), 401
        return jsonify(_bob_colab_runs.get(run_id, {"run_id": run_id, "status": "unknown"}))

    @app.post("/runs/<run_id>/cancel")
    def cancel(run_id):
        ok, error = _check_bearer_token(request.headers)
        if not ok: return jsonify({"error": error}), 401
        _bob_colab_cancelled.add(run_id)
        return jsonify({"run_id": run_id, "trace_id": run_id, "status": "cancelled"})

    return app

print(f"✅ Staged Bob Chat contract ready in {BOB_RUNTIME_MODE} mode with {BOB_MODEL_LANE_COUNT} lane(s)")
````

````python
# Phase 15 — Start and expose the Bob API from Lightning AI Studio
# MODE: BOTH — uses the Phase 0.2A selection
#
# Behaviour:
# - Asks for the ngrok token and Bob API token.
# - Press Enter to use the token already stored in environment secrets.
# - Starts or reuses the local Flask server.
# - Creates a fresh ngrok HTTPS tunnel.
# - Verifies both the local and public health endpoints.
# - Updates _colab_public_url for the rest of the notebook.

import os
import time
import requests
from getpass import getpass
from pyngrok import ngrok

PORT = int(os.environ.get("BOB_PORT", "8000"))


# ---------------------------------------------------------------------------
# 1. Request tokens, with environment-secret fallback
# ---------------------------------------------------------------------------

def get_token_with_secret_fallback(
    environment_name: str,
    display_name: str,
) -> str:
    """
    Ask the user for a token.

    Behaviour:
    - Enter a token: use the entered token.
    - Press Enter: use the token stored in os.environ.
    - No entered or stored token: raise a clear error.
    """
    stored_token = os.environ.get(environment_name, "").strip()

    if stored_token:
        prompt = (
            f"Enter {display_name} "
            f"[Press Enter to use stored secret]: "
        )
    else:
        prompt = (
            f"Enter {display_name} "
            f"[No stored secret found]: "
        )

    entered_token = getpass(prompt).strip()

    if entered_token:
        os.environ[environment_name] = entered_token
        print(f"✅ Using entered {display_name}.")
        return entered_token

    if stored_token:
        print(f"✅ Using stored {display_name} secret.")
        return stored_token

    raise RuntimeError(
        f"{display_name} was not entered and no stored "
        f"{environment_name} secret was found."
    )


NGROK_AUTH_TOKEN = get_token_with_secret_fallback(
    environment_name="NGROK_AUTH_TOKEN",
    display_name="ngrok auth token",
)

BOB_COLAB_TOKEN = get_token_with_secret_fallback(
    environment_name="BOB_COLAB_TOKEN",
    display_name="Bob API bearer token",
)

# Keep the selected values available to all later notebook cells.
os.environ["NGROK_AUTH_TOKEN"] = NGROK_AUTH_TOKEN
os.environ["BOB_COLAB_TOKEN"] = BOB_COLAB_TOKEN

print("✅ Authentication tokens configured.")
print("Token values were not displayed.")


# ---------------------------------------------------------------------------
# 2. Configure ngrok
# ---------------------------------------------------------------------------

ngrok.set_auth_token(NGROK_AUTH_TOKEN)

# Disconnect existing tunnels when possible.
try:
    existing_tunnels = ngrok.get_tunnels()

    for existing_tunnel in existing_tunnels:
        try:
            print(
                "Closing old ngrok tunnel:",
                existing_tunnel.public_url,
            )
            ngrok.disconnect(existing_tunnel.public_url)
        except Exception as disconnect_error:
            print(
                "⚠️ Could not disconnect tunnel cleanly:",
                disconnect_error,
            )

except Exception as list_error:
    print(
        "⚠️ Could not inspect existing ngrok tunnels:",
        list_error,
    )

# Stop the previous pyngrok process.
# This does not stop the Flask server thread.
try:
    ngrok.kill()
except Exception:
    pass

time.sleep(1)


# ---------------------------------------------------------------------------
# 3. Start or reuse the local Bob Flask server
# ---------------------------------------------------------------------------

if os.environ.get("BOB_PRELOAD_MODEL_LANES", "1") == "1":
    print(f"Preloading {BOB_MODEL_LANE_COUNT} lane(s) for {BOB_RUNTIME_MODE} mode sequentially...")
    preload_shared_model_lanes()

server_info = start_colab_http_server(
    port=PORT,
    use_ngrok=False,
)

if not isinstance(server_info, dict):
    server_info = {
        "port": PORT,
    }

LOCAL_BASE_URL = f"http://127.0.0.1:{PORT}"


def wait_for_local_server(
    base_url: str,
    attempts: int = 30,
    delay: float = 1.0,
):
    """
    Wait until the local Flask health endpoint responds.
    """
    last_error = None

    for attempt in range(1, attempts + 1):
        try:
            response = requests.get(
                f"{base_url}/health",
                timeout=5,
            )

            if response.ok:
                print(
                    f"✅ Local Bob API responded "
                    f"on attempt {attempt}."
                )
                return response

            last_error = RuntimeError(
                f"HTTP {response.status_code}: "
                f"{response.text[:500]}"
            )

        except Exception as exc:
            last_error = exc

        time.sleep(delay)

    raise RuntimeError(
        "The local Bob API did not become ready.\n"
        f"Health URL: {base_url}/health\n"
        f"Last error: {last_error}"
    )


local_health_response = wait_for_local_server(
    LOCAL_BASE_URL,
)


# ---------------------------------------------------------------------------
# 4. Create a fresh public ngrok HTTPS tunnel
# ---------------------------------------------------------------------------

_bob_ngrok_started = time.perf_counter()
try:
    tunnel = ngrok.connect(
        addr=PORT,
        proto="http",
        bind_tls=True,
    )

except Exception as exc:
    _bob_runtime_log("ngrok", "ngrok.tunnel_error", duration_ms=round((time.perf_counter() - _bob_ngrok_started) * 1000), error={"type": type(exc).__name__, "message": _safe_error_message(exc)})
    ngrok_log_text = ""

    try:
        ngrok_process = ngrok.get_ngrok_process()

        if ngrok_process and ngrok_process.proc:
            stdout = ngrok_process.proc.stdout

            if stdout:
                ngrok_log_text = stdout.read() or ""

    except Exception:
        pass

    raise RuntimeError(
        "ngrok could not create the public tunnel.\n"
        f"Original error: {type(exc).__name__}: {exc}\n"
        f"ngrok logs:\n{ngrok_log_text[-3000:]}"
    ) from exc


BASE_URL = str(
    getattr(tunnel, "public_url", "") or ""
).strip().rstrip("/")

if not BASE_URL:
    raise RuntimeError(
        "ngrok.connect() completed, but no public URL was returned."
    )

_bob_runtime_log("ngrok", "ngrok.tunnel_started", duration_ms=round((time.perf_counter() - _bob_ngrok_started) * 1000), public_host=_bob_urlparse.urlsplit(BASE_URL).hostname, port=PORT, tunnel_name=getattr(tunnel, "name", None), tls=BASE_URL.startswith("https://"))

if not BASE_URL.startswith("https://"):
    print(
        "⚠️ ngrok returned a non-HTTPS public URL:",
        BASE_URL,
    )


# ---------------------------------------------------------------------------
# 5. Synchronize notebook server state
# ---------------------------------------------------------------------------

# Update the global URL used by start_colab_http_server() and other cells.
globals()["_colab_public_url"] = BASE_URL

server_info.update(
    {
        "port": PORT,
        "local_url": LOCAL_BASE_URL,
        "public_url": BASE_URL,
        "tunnel_name": getattr(tunnel, "name", None),
    }
)


# ---------------------------------------------------------------------------
# 6. Verify the public ngrok URL
# ---------------------------------------------------------------------------

def wait_for_public_server(
    base_url: str,
    attempts: int = 20,
    delay: float = 1.5,
):
    """
    Wait until the Bob API is reachable through ngrok.
    """
    last_error = None

    for attempt in range(1, attempts + 1):
        try:
            response = requests.get(
                f"{base_url}/health",
                timeout=10,
            )

            if response.ok:
                print(
                    f"✅ Public Bob API responded "
                    f"on attempt {attempt}."
                )
                return response

            last_error = RuntimeError(
                f"HTTP {response.status_code}: "
                f"{response.text[:500]}"
            )

        except Exception as exc:
            last_error = exc

        time.sleep(delay)

    raise RuntimeError(
        "The ngrok URL was created, but its health endpoint "
        "did not become reachable.\n"
        f"Health URL: {base_url}/health\n"
        f"Last error: {last_error}"
    )


public_health_response = wait_for_public_server(
    BASE_URL,
)


# ---------------------------------------------------------------------------
# 7. Display connection information
# ---------------------------------------------------------------------------

print("\n" + "=" * 72)
print("✅ Bob model API is online")
print("=" * 72)

print("Local URL:       ", LOCAL_BASE_URL)
print("Public Base URL: ", BASE_URL)
print("Runtime mode:    ", BOB_RUNTIME_MODE)
print("Model lanes:     ", BOB_MODEL_LANE_COUNT)
print("Runtime logs:    ", BOB_RUNTIME_LOG_DIR)
print("Health URL:      ", f"{BASE_URL}/health")
print("Capabilities URL:", f"{BASE_URL}/capabilities")

print("\nBob IDE API paths:")
print("  Chat:         /chat")
print("  Plan:         /plan")
print("  Run:          /run-agent")
print("  Stream:       /run-agent/stream")
print("  Run status:   /runs/{run_id}")
print("  Cancel run:   /runs/{run_id}/cancel")

print("\nAuthentication:")
print("  Authorization: Bearer <BOB_COLAB_TOKEN>")
print("  Bearer token value was not printed.")

print("\nContract:")
print(" ", BOB_COLAB_CONTRACT_VERSION)

print("\nLocal health response:")
try:
    print(local_health_response.json())
except Exception:
    print(local_health_response.text[:1000])

print("\nPublic health response:")
try:
    print(public_health_response.json())
except Exception:
    print(public_health_response.text[:1000])

print("\nCopy this Base URL into Bob IDE:")
print(BASE_URL)

server_info
````
